</div>

<div class="alert alert-success" role="alert">

### Aufgabe 1.1: Planung und Dokumentation des Messvorhabens

</div>

🔧 **Hinweise zum Ablauf:** 🔧 

 * Die avisierte Messung ist vermeintlich einfach. Sie sollen Sie aber **frei und ohne weitere Vorgaben planen, dokumentieren und duchführen**.
 * Am Ende der Messung sollte **aus Ihrem Protokoll klar hervorgehen**, was Sie gemacht haben, um zu Ihrem Ergebnis zu gelangen. 
 * 🔔 Um dabei erfolgreich zu sein, sollten Sie die Planung der **Messreihe ernstnehmen** und die Messung daraufhin **mit aller gebotenen Sorgfalt durchführen und dokumentieren**.
 * 🔔 Ziel dieser Aufgabe ist es sich über die Planung der Messreihe bewusst zu werden und diese in Ihrem Protokoll festzuhalten. 
 * Wichtige Fragen, die Sie dabei beantworten sollten sind: 
   * Sollten Sie lieber ein kompaktes Metallstück benutzen oder ein Granulat?
   * Sollten Sie zur Herstellung der Mischtemperatur heißes Metall und kaltes Wasser verwenden oder umgekehrt? 
   * Ist Wasser oder eine andere Flüssigkeit geeigneter? 
   * Was sind günstige Anfangs- und Endtemperaturen für Ihre Messung? 
   * Welche Massen (für Wasser und Metall) sollten Sie benutzen und wie und zu welchem Zeitpunkt der Messung sollten Sie diese bestimmen? 
   * Sollten Sie Messfehler durch anhaftendes Wasser oder durch die an das Kalorimeter und Zubehör abgegebene Wärmemenge (Wärmegang) berücksichtigen? 
   * Wie wollen Sie den Wärmegang des gesamten Aufbaus abschätzen?
   * Wie schaffen Sie es bei Wiederholung der Messung immer wieder wohldefinierte Anfangsbedingungen herzustellen (das Metall erhitzt sich oder kühlt sich ab, das gleiche gilt für das Wasser oder das Kalorimeter)? 
   * Sind Ihre Messungen reproduzierbar und die Ergebnisse mit einer statistischen Streuung kompatibel, oder gibt es mögliche Tendenzen (werden die Ergebnisse z.B. mit zunehmender Länge der Messreihe tendenziell größer oder kleiner)?
   * Was ist ein geeignetes Thermometer in Hinblick auf Genauigkeit, Ablesegenauigkeit, oder störende eigene Masse? 
   * Genügt Ihnen jeweils eine Temperaturablesung oder sollten Sie die Messung der Temperatur in Abhängigkeit von der Zeit bestimmen?
   * Kann es sinnvoll sein Hilfsmessungen, mit bekannten Sollergebnissen (z.B. mit Wasser-Wasser-Gemischen) vorzunehmen aus denen Sie die Mischtemperatur vorhersagen können?

Beschreiben und begründen Sie Ihre Versuchsplanung, **messen Sie sorgfältig und dokumentieren Sie Ihr Vorgehen genau.**

## Aufgabe 2: Spezifische Wärmekapazität von Aluminium als Funktion der Temperatur

</div>

<div class="alert alert-success" role="alert">

### Aufgabe 2.1: Datennahme

</div>

🔧 **Hinweise zum Ablauf:** 🔧 

 * Bei dieser Aufgabe ist der **Ablauf der Messreihe vorgegeben**. Zusätzlich stellen wir Ihnen **zwei Kontrollmessungen** zur Verfügung, die Sie im zeitlichen Rahmen des Praktikums nicht selbst durchführen können. 
 * Studieren Sie die Grundlagen dieses Versuchs [hier](https://gitlab.kit.edu/kit/etp-lehre/p2-praktikum/students/-/blob/main/Spezifische_Waermekapazitaet/doc/Hinweise-Waermekapazitaet-Verlauf.md), damit Sie sich **darüber im Klaren sind, was Sie bei diesem Versuch tun**, was das Ziel Ihrer Messung ist und an welchen Stellen besondere Sorgfalt von Ihnen verlangt wird.  
 * Kühlen Sie den $\mathrm{Al}$-Hohlzylinder (AL) mit Flüssigstickstoff im Nalgene-Isolierbehälter auf $T=-196^{\circ}\mathrm{C}$ (Siedetemperatur von Flüssigstickstoff) ab. 
 * 💡 Bereiten Sie den Edelstahl-Isolierbehälter (EI) durch Ausschwenken mit Flüssigstickstoff auf die Aufnahme des AL vor. 💡 Warum?
 * Während der Messreihe heizen Sie den AL mit Hilfe der eingebauten Heizwicklung innerhalb des EI mit einer konstanten Heizleistung (von $\lesssim 20\,\mathrm{W}$) wieder auf und zeichnen den Verlauf der Temperatur als Funktion der Zeit $T(t)$ mit Hilfe eines Datenloggers auf.
 * ⚠️ **Heizen Sie den AL nicht mit einer Stromstärke $I>2\ \mathrm{A}$, da sie sonst Gefahr laufen den Aufbau zu zerstören.** ⚠️
 * Zum Zweck der Temperaturmessung ist ein $\text{NiCr-Ni}$-Thermoelement (TE) im AL befestigt.
   * Mit Hilfe von Eiswasser können Sie dem TE eine stabile Referenztemperatur von $\vartheta_{\mathrm{ref}}=0^{\circ}\mathrm{C}$ zur Verfügung stellen. 💡 Warum $\vartheta_{\mathrm{ref}}=0^{\circ}\mathrm{C}$?
   * 🔔 Achten Sie dabei darauf, verbleibende Eisstücke fein genug zu zerdrücken und rühren Sie den Eiswasserbehälter gelegentlich um, um eine homogene Temperaturverteilung um $0^{\circ}\mathrm{C}$ zu garantieren.
   * 🔔 Es empfiehlt sich die Temperatur des Eiswassers mit Hilfe eines normalen Thermometers ebenfalls zu kontrollieren.
   * Sie müssen das TE für die Messung von Temperaturen erst noch geeignet kalibrieren, d.h. die vom TE ausgegebenen Spannungen in Temperaturen umrechnen (siehe **Aufgabe 2.2**).
