## Aufgabe 1: Grundlagen

### Aufgabe 1.1: Photoelektrischer Effekt (Demonstrationsversuch)

**Hinweise zum Ablauf:**

 * Bei diesem Versuchsteil handelt es sich um einen **Demonstrationsversuch, den alle Gruppen gemeinsam mit Ihrem:r Tutor:in durchführen**.
 * Dabei sind **Sorgfalt und Experimentiergeschick** gefragt.
 * Gehen Sie bei diesem Versuch wie folgt vor:
   * Laden Sie Zn-Platte **positiv** auf und beobachten Sie das Elektrometer, wenn Sie die Zn-Platte mit der Hg-Dampflampe bestrahlen.
   * Laden Sie die Zn-Platte **negativ** auf und beobachten Sie das Elektrometer, wenn Sie die Zn-Platte mit der Hg-Dampflampe bestrahlen.
   * Laden Sie die Zn-Platte negativ auf und beobachten Sie das Elektrometer, wenn Sie die Zn-Platte mit der Hg-Dampflampe bestrahlen und **zusätzlich eine positiv geladene Elektrode** in die Nähe der Zn-Platte bringen.
 * Beachten Sie dabei die folgenden Hinweise:
   * Eine sich mit der Zeit bildende Oxydschicht kann die Photoemission der Zn-Platte stark beeinträchtigen. **Polieren Sie Zn daher vor Versuchsbeginn sorgfältig**.
   * Laden Sie die Zn-Platte daraufhin mit einer Spannung von etwa $\pm2\,\mathrm{kV}$ auf. Sie können hierfür die grobe Skala auf der Hochspannungsquelle nutzen. Es geht in erster Linie darum, dass Sie **Ladung auf die Zn-Platte aufbringen**.
   * Der Ausschlag des Versoriums ist träge; er ändert sich daher nur sehr langsam. Es bietet sich an, dass Sie die Spannung am Drehregler **langsam nach oben regeln**, bis Sie mit dem Ausschlag des Versoriums zufrieden sind.
   * Erfahrungsgemäß hält die Zn-Platte die aufgebrachte Ladung leicht für mehrere Minuten. In der Praxis wird sich die Zn-Platte jedoch aufgrund von Wechselwirkungen mit der Umgebung mit zunehmender Zeit entladen. Wie schnell dies geschieht hängt von äußeren Bedingungen, wie der Luftfeuchte ab. Die Halbwertszeit für diesen Vorgang können Sie mit Hilfe einer Messung ohne Einstrahlung der Hg-Lampe grob abschätzen.

**Hinweise zur Auswertung:**
 
 * Beschreiben Sie Ihre Beobachtungen.
 * Es handelt sich, bei dem was Sie beobachten um ein Quantenphänomen. Sie können den Vorgang, wenn Sie möchten, mit Ihrem Mobiltelefon aufzeichnen.
 * Erklären Sie alle Beobachtungen, die Sie bei diesem Versuch machen. Beziehen Sie hierzu die folgenden Fragen in Ihre Diskussion ein:
   * Wie groß ist die **Austrittsarbeit $E_{A}$** eines Elektrons aus Zn (in $\mathrm{eV}$)?
   * Welcher **Wellenlänge $\lambda_{A}$ des Lichts** entspricht dies (in $\mathrm{nm}$)?
   * Wie vergleicht sich $\lambda_{A}$ mit dem [Spektrum der Hg-Lampe](https://de.wikipedia.org/wiki/Quecksilberdampflampe)?
   * Wie sieht die Transmission bzw. Absorption von Licht in (gewöhnlichem) Glas aus?
   * Welche Konsequenzen hat dies für den Versuchsaufbau?
   * Kann man diesen Versuch in einem geschlossenen Raum bei Tageslicht durchführen oder sollte man verdunkeln?

## Aufgabe 2: Bestimmung von $h$

### Aufgabe 2.1: Ladekurve als Funktion der Frequenz

**Hinweise zum Ablauf:**

 * Beachten Sie, dass die Hochdruck-Hg-Dampflampe ihr **Intensitätsmaximum erst ${\approx}10\,\mathrm{min}$ nach Inbetriebnahme** erreicht. 
 * Achten Sie auf eine **zentrale und gute Ausleuchtung der Photozelle**.
 * Beachten Sie, dass die **ringförmige Anode** nicht durch das Licht der Hg-Dampflampe getroffen wird.
 * Schließen Sie über die Messanordnung einen geeigneten Kondensator parallel zum Signal der Photodiode.
   * Wir empfehlen mit einer Kapazität von $4.7\,\mathrm{nF}$ zu beginnen.
   * Es kann sich allerdings anbieten, je nach Wellenlänge diesen Kondensator gegen einen Kondensator größerer oder geringerer Kapazität auszutauschen.
   * Notieren Sie sich in diesem Fall **für welche Messreihen Sie welche Kondendatoren** verwendet haben.    
 * Schalten Sie Schritt für Schritt einen Filter nach dem anderen vor den Lichtschutzkollimator der Photozelle und zeichnen Sie die Ladekurve des Kondensators auf. Die Filter sind aufsteigend in $\lambda_{\mathrm{CWL}}^{(i)}$ montiert.
 * Je nach Wahl des Kondensators empfihelt sich eine Messteihe zwischen $5-10\,\mathrm{min}$. Sie sollten das Plateau der Messung einigermaßen klar erkennen können. 

**Hinweise zur Auswertung:**

 * Beschreiben Sie Ihr Vorgehen bei der Messung.
 * Stellen Sie alle Messreihen in einem **gemeinsamen Diagramm** mit entsprechenden Labels oder entsprechender Legende dar.
 * Wählen Sie die Ladekurve **für eine Wellenlänge** aus und passen Sie daran ein Modell sowohl nach Gleichung **(3)** als auch nach Gleichung **(4)** [hier](https://gitlab.kit.edu/kit/etp-lehre/p2-praktikum/students/-/blob/main/Photoeffekt/doc/Hinweise-Photoeffekt.md) an.
   * Geben Sie jeweils den $\chi^{2}$-Wert der Anpassung geteilt durch die Anzahl der Freiheitsgrade des Modells an und diskutieren, wie gut das jeweilige Modell die Daten beschrieben kann.
   * **Wählen Sie das bessere Modell zur Beschriebung der Daten für alle Wellenlängen** aus und bestimmen Sie daraus $U_{\mathrm{Ph}}^{(i)}\pm\Delta U_{\mathrm{Ph}}^{(i)}$.
 * Tragen Sie die gewonnen Werte für $U_{\mathrm{Ph}}\pm\Delta U_{\mathrm{Ph}}$ gegen $\nu_{\mathrm{CWL}}\pm\Delta\nu_{\mathrm{CWL}}$ auf.
 * Passen ein Modell nach Gleichung **(2)** [hier](https://gitlab.kit.edu/kit/etp-lehre/p2-praktikum/students/-/blob/main/Photoeffekt/doc/Hinweise-Photoeffekt.md) an den Verlauf der sich ergebenden Datenpunkte an.
   * Geben Sie den $\chi^{2}$-Wert der Anpassung geteilt durch die Anzahl der Freiheitsgrade des Modells an und diskutieren, wie gut das Modell die Daten beschreiben kann.
   * Beachten Sie, aus Gründen der **numersichen Stabilität der Anpassung** darauf, dass bei der Übergabe an den Algorithmus zur Anpassung für die Bestimmung von $h$, sowohl $U_{\mathrm{Ph}}^{(i)}$, als auch $\nu_{\mathrm{CWL}}^{(i)}$ numerische Werte $\mathcal{O}(1)$ annehmen. Sie erreichen dies z.B. indem Sie $U_{\mathrm{Ph}}^{(i)}$ in $\mathrm{eV}$ und $\nu_{\mathrm{CWL}}^{(i)}$ in $\mathrm{THz}$ angeben. 
   * Sowohl [$e=1.602176634\times10^{-19}\ \mathrm{C}$](https://de.wikipedia.org/wiki/Elementarladung) als auch [$c=2.99792458\times10^{8}\ \mathrm{m/s}$](https://de.wikipedia.org/wiki/Lichtgeschwindigkeit) sind SI-Einheiten definierende **Naturkonstanten, die ohne Unsicherheiten definiert sind**. Sie können diesen Umstand später berücksichtigen, um $h$ direkt zu bestimmen.
