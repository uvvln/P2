Dieser Versuch wurde am Versuchsaufbau 153 durchgeführt. Da dort am RZ noch eine
Gasentladungsröhre eingebaut war.
Für diesen Versuch wurden die Ventile V1 und V2 geöffnet und somit die DSP mit dem RZ
verbunden.
Zum Evakuieren wurde hier ausschließlich die DSP verwendet, weshalb die Gasentladung nur
in einem tendenziell gröberem Vakuum untersucht wurde.
Die Gasentladungsröhre wurde, zuerst noch bei Normaldruck, unter Spannung gesetzt.
Innerhalb der Röhre wurden damit Elektronen beschleunigt, welche mit der in der Röhre
befindenden Luft, größtenteils Stickstoff, stießen.
Haben die Elektronen genug Energie, also haben sie eine gewissen Strecke im
Beschleunigungsfeld zurückgelegt, so können die Stickstoffatome kurzzeitig angeregt
werden, woraufhin sie sich wieder abregen und ein Photon mit einer charakteristischen
Wellenlänge emittieren.
Nun werden die Elektronen weiter beschleunigt, bis sich der Vorgang in äquidistanten
Abständen wiederholt.
Dies geschieht bei Normaldruck allerdings nur kaum bis gar nicht, da die Elektronen zu oft
durch Stöße an Energie verlieren, bevor sie zu einer Anregung genug Energie haben.
Wird der Druck allerdings gesenkt, so findet zuerst immer mehr Gasentladung statt, da nun
weniger Stoßpartner vorhanden sind, die die Elektronen vorzeitig entschleunigen. Somit wird
das beobachtbare Licht erst immer heller.
In extremen Tiefdruckbereichen lässt die beobachtbare Intensität jedoch irgendwann nach,
da mit der fortlaufenden Evakuierung die Wahrscheinlichkeit sinkt, mit der die Elektronen
überhaupt einen Stoßpartner treffen

### Aufgabe 2.1: Saugvermögen der DSP

**V E R S U C H S B E S C H R E I B U N G**

In diesem Versuch sollten wir das Saugvermögen der Drehschieberpumpe (DSP) messen. Hierfür wurde die Apparatur vollständig belüftet. Anschließend wird der Druck mithilfe der Drehschieberpumpe verringert und mit der Messstation T1 für einen Zeitraum von 5 Minuten alle 5 Sekunden ein Wert gemessen. Mithilfe dieser Messwerte wird nun das Saugvermögen S(p) berechnet, das Volumen konnten wir aus dem Datenblatt entnehmen, für n wählten wir den Wert $1$.

$$ 
S(p) =  \frac{V_{RZ}}{n(t_i + 1 - t_i)}  ln(\frac{p_i}{pi+ 1})

```python
p_21 = np.array([1000, 1000, 1000, 1000, 1000, 400, 150, 80, 51, 30, 20, 13.5, 9.13, 6.25, 4.27, 2.91, 2.04, 1.5, 1.14, 0.72, 0.69, 0.48, 0.37, 0.30, 0.25, 0.21, 0.185, 0.164, 0.149, 0.134, 0.125, 0.114, 0.106, 0.1, 0.0942, 0.0903, 0.0871, 0.0832, 0.08, 0.078, 0.076, 0.075, 0.073, 0.071, 0.069, 0.067, 0.065, 0.064, 0.063, 0.062, 0.061, 0.060, 0.059, 0.058, 0.057, 0.056, 0.055, 0.055, 0.054, 0.053])

t=np.arange(0, len(p_21) * 5, 5)

#Unsicherheiten
delta_p = np.zeros_like(p_21)
for i, p in enumerate(p_21):
    if 5e-4 <= p < 1e-3:
        delta_p[i] = 0.10 * p  
    elif 1e-3 <= p < 1e2:
        delta_p[i] = 0.02 * p  
    elif 1e2 <= p <= 1e3:
        delta_p[i] = 0.25 * p  
    else:
        delta_p[i] = 0.02 * p 
        

plt.xscale('log')
plt.errorbar(t, p_21, yerr=delta_p, xerr=0.0539, fmt='x', ecolor='blue', capsize=2)
plt.xlabel('Zeit (s)')
plt.ylabel('Druck (mbar)')
plt.title('Druck-Zeit Diagramm')
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
V = 10.119  

def S_p(t,p,p_0,V,t_0 = 0):
    return V/(t_0-t)*np.log(p/p_0)

def S_err(t,p,p_0,V,t_0=0):
    return V*np.sqrt(((0.0539)/(t-t_0))**2+((0.196*np.log(p/p_0))/(t-t_0)**2)**2)

def mean(list,start, finish):
    sum = 0
    i = 0
    for _ in range(start,finish+1):
        sum += list[_]
        i += 1
    return sum/i

S = [S_p(t[i]+20,p_21[i],p_0=p_21[0],V=V) for i in range(len(p_21))]
Serr= [S_err(t[i]+20,p_21[i],p_0=p_21[0],V=V) for i in range(len(p_21))]

plt.errorbar(x=p_21,y=S,xerr=0.0539*p_21,yerr=Serr, fmt='x', ls='none')
plt.title('Saugvermögen')
plt.xlabel('Druck in $mbar$')
plt.ylabel(r'S in $\frac{{l}}{{s}}$')
plt.xscale('log')
plt.ylim(-0.2,1.2)
plt.axvline(x=p_21[9],color="red")
plt.axvline(x=p_21[23],color="red")
plt.axhline(y=mean(list=S,start=9,finish=23),xmin=0.21,xmax=0.63,color='orange')
plt.text(x=p_21[7],y=1,s='Bereich I',fontweight='bold',fontsize='medium')
plt.text(x=p_21[18],y=1,s='Bereich II',fontweight='bold',fontsize='medium')
plt.text(x=p_21[59],y=1,s='Bereich III',fontweight='bold',fontsize='medium')

plt.show()
```

```python
plt.errorbar(t[9:23],p_21[9:23],yerr=0.0539*p_21[9:23],xerr=0.196, fmt='x', ls='none')
plt.xscale('log')
plt.title('Druck-Zeit Diagramm Bereich II')
plt.xlabel(('Zeit in s'))
plt.ylabel('Druck in mbar')
plt.show()
```

```python
def moddel_21(t,t_0=t[9],p_0=p_21[9],S=mean(list=S,start=9,finish=23)):
    return np.log(p_0*np.exp(-(S)/(V)*(t-t_0)))

dataset = kafe2.XYContainer(x_data=t[9:23],y_data=np.log(p_21[9:23]))
dataset.axis_labels=['Zeit in s', 'log(p) in mbar']
dataset.label='Druck-Zeit Diagramm Bereich II'
dataset.add_error(axis='x',err_val=0.196)
dataset.add_error(axis='y',err_val=0.0539,relative=True)

fit = kafe2.Fit(dataset,model_function=moddel_21)
fit.limit_parameter("t_0", lower=0)

fit.do_fit()
fit_result=kafe2.Plot([fit])

fit_result.plot()
plt.show()
```

**D I S K U S S I O N**

Schaut man sich nun den zweiten Graphen an so erkennt man das sich dieser in drei Bereiche einteilen lässt. Im ersten Bereich steigt das Saugvermögen an da anfangs die Gasdichte sehr hoch ist und die Pumpe ineffizent arbeitet. Im zweiten Bereich ist das Saugvermögen Konstant hier arbeitet die Pumpe am effektifsten. Im dritten Bereich kommt die DSP an ihre grenzen und die Gasdichte wird sehr gering und sorgt somit dafür das nur noch wenige Teilen in die Pumpe kommen. Schaut man sich nun den zweiten Bereich von Sekunde $45$ bis $110$ genauer an und führt diesen Fit durch $p(t)=p_0 ⋅ e^{-\frac {S}V(t−t_0)}$, so sieht man das vorallem am Anfang des zweiten Bereichs eine große Abweichung herscht. ### Aufgabe 3.1: Statische Kalibration von T3

Führen Sie mit Hilfe des Referenzvolumes RV eine statische Kalibration des Vakuummeters T3 durch.

</div>

**V E R S U C H S B E S C H R E I B U N G**

Zur Kalibration des Vakuummeters T3 wird ein einstufiges statisches Kalibrationsverfahren mit dem Referenzvolumen $RV$ durchgeführt. Grundlage ist das Boyle-Mariotte-Gesetz.

Zunächst wird der Rezipient evakuiert und durch Schließen von V2 vom restlichen System getrennt. Anschließend wird durch wiederholtes Öffnen von V3 ein Druckausgleich zwischen Rezipient und Referenzvolumen hergestellt. Der dabei von T3 angezeigte Druck $p_{\mathrm{RZ}}^{\mathrm{(T3)}}$ wird nach Erreichen des Gleichgewichtszustands aufgenommen. Danach wird V3 wieder geschlossen und über B2 erneut Atmosphärendruck im Referenzvolumen eingestellt. Dieser Vorgang wird bis zu einem Druck von etwa $80\,\mathrm{mbar}$ wiederholt.

Die Druckänderung ergibt sich aus

$$
\Delta p_i
=
\frac{V_{\mathrm{RV}}}{V_{\mathrm{RV}}+V_{\mathrm{RZ}}}
\left(p_0-p_{\mathrm{RZ},i}\right).
$$

Mit

$$
p_{\mathrm{RZ},i+1}
=
p_{\mathrm{RZ},i}+\Delta p_i
$$

werden die tatsächlichen Druckwerte rekursiv bestimmt. Aus den Datenpunkten

$$
\left(
p_{\mathrm{RZ},i},
p_{\mathrm{RZ},i}^{\mathrm{(T3)}}
\right)
$$

wird anschließend die Kalibrationskurve

$$
K\left(p_{\mathrm{RZ}}, p_{\mathrm{RZ}}^{\mathrm{(T3)}}\right)
$$

des Vakuummeters T3 erstellt.

---


Zunächst wird ein linearer Fit, dann ein Fit zu einem Polynom höherer Ordnung gemacht. Die Unsicherheiten der aufgenommenen Messwerte $p_{\mathrm{RZ},i}^{\mathrm{(T3)}}$ wurden dabei als 0.5 % angenommen. Dieser Fehler wurde entsprechend der Auflösung der Anzeige gewählt.

```python
import numpy as np
import matplotlib.pyplot as plt
from uncertainties import ufloat
import uncertainties.unumpy as unp
from kafe2 import Fit, MultiFit, Plot, HistFit, HistContainer, XYContainer


uV_RV = ufloat(0.034, 0)#in l
uV_RZ = ufloat(10.8, 0.3) #in l
up_0 = ufloat(1000,1) #in mbar

p_RZ_T3 = np.array([0.285, 4.05, 8.48, 13.1, 18.3, 23.8, 29.5, 36.1, 43.7, 51.0, 59.8, 70.0, 80.1,96.6, 115.0, 134, 153, 176, 217, 251])
up_RZ_T3 = unp.uarray(p_RZ_T3, p_RZ_T3*0.005)


ufaktor = uV_RV / (uV_RV + uV_RZ)

# Array für berechnete reale Drücke
up_RZ = unp.uarray(np.zeros(len(up_RZ_T3)), 0)

# Startwert
up_RZ[0] = up_RZ_T3[0]

# Rekursive Berechnung
for i in range(len(up_RZ) - 1):
    delta_p = ufaktor * (up_0 - up_RZ[i])
    up_RZ[i + 1] = up_RZ[i] + delta_p

# modelle
def polynom(x, a, b, c, d):
    return a*x**b + c*x+d

def linear(x, a, b):
    return a*x + b

# cutte bei 80 mbar

up_RZ_T3_cut = up_RZ_T3[0:13]
up_RZ_cut = up_RZ[0:13]

xydata = XYContainer( unp.nominal_values(up_RZ_cut), unp.nominal_values(up_RZ_T3_cut))
xydata.add_error(axis='x', err_val=unp.std_devs(up_RZ_cut))
xydata.add_error(axis= 'y', err_val= unp.std_devs(up_RZ_T3_cut))
xydata.label= "Kalibrationskurve von T3"

fit = Fit(data=xydata, model_function=linear)

# do the fit
fit.do_fit()
plot = Plot(fit)

# data
plot.x_label = r"$p_{\mathrm{RZ}}$ [mbar]"
plot.y_label = r"$p_{\mathrm{RZ}}^{(T3)}$ [mbar]"
plot.plot()
plt.show()

# create the Fit object by specifying a density function
fit = Fit(data=xydata, model_function=polynom)

# do the fit
fit.do_fit()
plot = Plot(fit)

# data
plot.x_label = r"$p_{\mathrm{RZ}}$ [mbar]"
plot.y_label = r"$p_{\mathrm{RZ}}^{(T3)}$  [mbar]"
plot.plot()
plt.show()
```

Die Tupel $\left(p_{\mathrm{RZ},i},p_{\mathrm{RZ},i}^{\mathrm{(T3)}}\right)$ wurden wie oben erklärt bestimmt und auf zwei Modelle gefittet. Das lineare Modell zeigt gerade bei höher werdenden Drücken eine geringe Übereinstimmung mit den Messdaten. Das kann man auch an dem sehr hohen $\chi^2/\alpha$ - Wert von 41.82 sehen.
Das zweite Modell hat zusätzlich zu dem linearen Anteil noch einen Term höherer Ordnung:

$$
f(x) = a \cdot x^b + c \cdot x + d.
$$

Die Pameter für dieses Modell wurden als $a = 0.13 \pm 0.01$, $b = 2.19 \pm 0.21$, $c = 1.160 \pm 0.058$ und $d = -0.047 \pm 0.016$ bestimmt. Das Modell passt sehr gut zu den Datenpunkten, was man an dem sehr kleinen $\chi^2/\alpha$ - Wert von 0.1869 sehen kann. Dieser Wert ist viel kleiner als 1 und damit passt der Fit eher zu gut zu den Daten. Dies spricht für zu konservativ angenommene Unsicherheiten.

Da die Datenpunkte deutlich von dem idealen Modell $p_{\mathrm{RZ}}^{\mathrm{(T3)}}(p_{\mathrm{RZ}}) = \alpha\,p_{\mathrm{RZ}}+\beta$ (mit $\beta$ = 0 und $\alpha$ = 1) abweichen, kann man gerade bei hohen Drücken von einer ungenügenden Kalibration des Wärmeleitungs-Vakuummeters ausgehen.

In diesem Versuch wird die elektrische Durchschlagfestigkeit der Kugelelektroden (KE) in Luft als Funktion des Umgebungsdrucks untersucht. Dazu wird die Spannung $U_{\mathrm{KE}}$ bestimmt, bei der zwischen den Elektroden eine Gasentladung beziehungsweise ein Plasma auftritt.

Zunächst wird der Rezipient ausschließlich mit der Drehschieberpumpe (DSP) evakuiert, während die Turbomolekularpumpe (TMP) ausgeschaltet bleibt. Die Messung beginnt bei Atmosphärendruck. Für jeden Messpunkt wird nach Erreichen des gewünschten Drucks das Ventil V1 geschlossen, damit der Druck im Rezipienten konstant bleibt. Anschließend wird die Spannung zwischen den Kugelelektroden schrittweise erhöht, bis eine Entladung beobachtet wird. Der zugehörige Druck sowie die Durchbruchsspannung werden notiert.

Der Druck wird danach schrittweise weiter abgesenkt, zunächst bis etwa

$$
p \approx 0.08\,\mathrm{mbar}.
$$

Um zusätzliche Messpunkte bei kleineren Drücken zu erhalten, wird anschließend auch die TMP verwendet, bis der minimale erreichbare Druck von ungefähr

$$
p \lesssim 10^{-4}\,\mathrm{mbar}
$$

erreicht ist. Danach wird die TMP abgeschaltet und V2 geschlossen. Durch den nun langsam ansteigenden Druck können weitere Messwerte aufgenommen werden.

Insgesamt wurden 19 Datenpunkte der Form

$$
\left(U_{\mathrm{KE}},\,p_{\mathrm{RZ}}\right)
$$

bestimmt werden. Spannungen werden mit einem Multimeter gemessen.

---

Wie im vorherigen Versuch werden Drücke p mit einer Unsicherheit von 0.5 % versehen. Da die Spannungen während der Messung stark fluktuiert haben, wird hier von einer Unsicherheit von 5 % ausgegangen.

```python
import numpy as np
U1 = np.array([4.2, 1.1, 1.0, 0.84, 0.68, 0.57, 0.43, 0.38, 0.39, 0.36, 0.395, 0.49, 0.58, 0.51, 0.48])#in kV
p1 = np.array([1000, 430, 258, 135, 70, 40, 17, 9, 4.9, 2.5, 1.3, 0.56, 0.28, 0.16, 0.12])#in mbar

U2 = np.array([1.7, 0.97, 0.58, 0.48])#in kV
p2 = np.array([0.001, 0.002, 0.009, 0.07])# in mbar

U = np.concatenate((U1, U2))
p = np.concatenate((p1, p2))

plt.figure()
plt.errorbar(p, U, xerr=0.005*p, yerr=0.05*U, fmt='o')
plt.xlabel('Druck (mbar)')
plt.ylabel('Spannung (kV)')     

plt.xscale("log")
plt.title('Durchschlagsfeldstärke in Luft')
plt.grid()
plt.show()
```

Die aufgenommenen Spannungen entsprechen jeweils dem Maximalwert unmittelbar vor der Entladung. Sobald zwischen den Kugeln ein Ladungsausgleich durch Funkenüberschlag oder bei niedrigen Drücken durch eine Glimmentladung stattfindet, sinkt die Spannung leicht ab, da nun ein Strom fließt.

Zur besseren Darstellung des Bereichs minimaler Überschlagsspannung wurde eine logarithmische x-Achse gewählt. Es ist erkennbar, dass die Überschlagsspannung bei sehr kleinen Drücken wieder ansteigt. Anfangs lag sie sogar oberhalb der maximal verfügbaren Spannung von $9\,\mathrm{kV}$ und konnte daher nicht gemessen werden.

Liegt die mittlere freie Weglänge der Gasteilchen in der Größenordnung des Elektrodenabstands, so wird die Überschlagsspannung besonders klein. Wird der Druck weiter erhöht, verkleinert sich die mittlere freie Weglänge erneut, wodurch die Durchschlagfestigkeit wieder zunimmt.

Dieses Verhalten kann man gut im Graph oben ablesen. Im Bereich 0.01 - 10 mbar bildet sich ein Plateau als Minimum der Überschlagsspannung.
