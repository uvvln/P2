### Aufgabe 3.1: Statische Kalibration von T3

</div>

🔧 **Hinweise zum Ablauf:** 🔧

 * Gehen Sie zur Erstellung der Kalibrationskurve $K\left(p_{\mathrm{RZ}}, p_{\mathrm{RZ}}^{\mathrm{(T3)}}\right)$ vor wie [hier](https://gitlab.kit.edu/kit/etp-lehre/p2-praktikum/students/-/blob/main/Vakuum/doc/Hinweise-Vakuummeter.md) beschrieben.
 * Beachten Sie zur Definition der Variablen die Diskussion [hier](https://gitlab.kit.edu/kit/etp-lehre/p2-praktikum/students/-/blob/main/Vakuum/doc/Hinweise-Vakuummeter.md).
 * Sie sollten auf diese Weise 15--20 Messpunkte für $p_{\mathrm{RZ},i}^{\mathrm{(T3)}}$ erhalten können. 
 * 🔔 **Um eine gute Messung zu erhalten ist es essentiell, beim Öffnen und Schließen der Ventile jeweils den Druckausgleich im Gleichgewichtszustand abzuwarten**. 

---

🖥️ **Hinweise zur Auswertung:** 🖥️
 
 * ✍🏻 **Protokollieren Sie:** ✍🏻 
    * Tragen Sie die Werte $p_{\mathrm{RZ},i}^{\mathrm{(T3)}}\pm\Delta p_{\mathrm{RZ},i}^{\mathrm{(T3)}}$ gegen die zugehörigen Werte $p_{\mathrm{RZ},i}\pm\Delta p_{\mathrm{RZ},i}$ auf.
    * Schätzen Sie hierzu geeignete Unsicherheiten $\Delta p_{\mathrm{RZ},i}^{\mathrm{(T3)}}$ und $\Delta p_{\mathrm{RZ},i}$ ab.
    * Passen Sie ein lineares Modell $p_{\mathrm{RZ}}^{\mathrm{(T3)}}(p_{\mathrm{RZ}}) = \alpha\,p_{\mathrm{RZ}}+\beta$ an die Datenpunkte an.
    * 🔔 **Beurteilen Sie die Kalibration und Linearität von T3** auf Grundlage des auf die Freiheitsgrade des Modells normierten $\chi^{2}$-Werts der Anpassung.  
    * Passen Sie an die Datenpunkte ein beliebiges **Polynom höherer Ordnung** an.
       * Fügen Sie Ihrer Auswertung den auf die Freiheitsgrade des Modells normierten $\chi^{2}$-Wert dieser Anpassung zu.
       * Diskutieren Sie die Signifikanz von Modellparametern, die zu nicht-linearen Termen dieses Modells gehören.
    * ⚠️Beachten Sie: Jede statistisch signifikante Abweichung von einem linearen Modell mit $\alpha=1,\,\beta=0$ weißt auf eine schlechte Kalibration der Anzeige von T3 hin.
    * Fügen Sie Ihrer Auswertung eine Diskussion unter diesem Aspekt zu.
    * 💡 Abschließend sollten Sie, falls Sie es für nötig erachten, die Messkurven aus **Aufgabe 2.3**, die Messungen bei T3 beinhalteten auf **Grundlage der Neukalibration neu auswerten**.
       * Dies für jede Kurve zu tun steht Ihnen frei. 
       * Sie sollten aber für mindestens eine Kurve den Effekt der Neukalibration demonstrieren, indem Sie die neu kalibrierte und die ursprüngliche Kurve gemeinsam darstellen.
       * Zur Demonstration des Effekts bietet sich auch eine zusätzliche Darstellung an, in der Sie das Verhältnis der beiden Kurven angeben.

### Aufgabe 3.2: Elektrische Durchschlagfestigkeit

</div>

🦉 **Hintergrundinformation zur elektrischen Durchschlagfestigkeit von Gasen finden Sie z.B. unter den folgenden Links:** 🦉

🔧 **Hinweise zum Ablauf:** 🔧

Die elektrische Durchschlagfestigkeit quantifizieren wir durch die **Spannung $U_{\mathrm{KE}}$, ab der es zwischen den KE zu elektrischen Entladungen kommt**. Gehen Sie hierzu wie folgt vor: 

 * Schalten Sie die TMP zunächst aus und evakuieren Sie den RZ **nur mit Hilfe der DSP**. 
 * 💡 **Schließen Sie vor der Bestimmung jedes Messpunkts nach erreichen des gewünschten Drucks V1**, so dass der Druck im RZ während der Messung konstant bleibt.
 * 💡 **Beginnen Sie mit der belüfteten Apparatur** bei Atmosphärendruck.
 * Erhöhen Sie die Spannung zwischen den KE bis es zur Gasentladung kommt.
 * Evakuieren Sie daraufhin den RZ bis auf **halben Atmosphärendruck** und erhöhen Sie die Spannung wieder bis zur Entladung. 
 * Wiederholen Sie diese Vorgehensweise bis Sie einen Druck von $p\approx0.08\ \mathrm{mbar}$ erreicht haben. In diesem Druckbereich wird es zunehmend schwieriger den Druck im RZ konstant zu halten.
 * 💡 Um leichter und schneller an weitere Messwerte zu gelangen, evakuieren Sie nun zusätzlich und ohne Unterbrechung mit der TMP bis zum minimal erreichbaren Druck (von $p\lesssim10^{-4}\ \mathrm{mbar}$).
 * Schalten Sie die TMP aus und schließen Sie V2. Der Druck im RZ wird nun von selbst steigen. 
 * Nehmen Sie sobald wie möglich (für $U \leq9\ \mathrm{kV}$) weitere Messwerte auf. Die Messreihe endet, wenn ein Druck von $p\approx0.08\ \mathrm{mbar}$ erreicht ist.
 * Auf diese Weise sollten Sie **15--20 Datenpunkte** $(U_{\mathrm{KE}},p_{\mathrm{RZ}})$ aufnehmen können.
 * ⚠️ **Beachten Sie:** ⚠️
    * Bei Atmosphärendruck sollten Sie eine klassische Gasentladung beobachten können.
    * Ab dem Grobvakuum ist jedoch vermehrt nur noch ein Plasma zu beobachten, dass nahe der Zündspannung fast gar nicht zu erkennen ist.
    * Die Messungen sind daher eher qualitativer Natur. 🔔 **Wichtig für diese Messung ist, dass Sie mit abnehmendem Druck ein Minimum bei der Durchschlagfestigkeit beobachten sollten. Diese sollte mit weiter fallendem Druck an wieder ansteigen.** 🔔
    * Mit was für einer Messanordnung würde man den Spannungsdurchbruch besser quantitativ bestimmen können?

---

🖥️ **Hinweise zur Auswertung:** 🖥️
 
 * ✍🏻 **Protokollieren Sie:** ✍🏻

### Aufgabe 3.3: Aufdampfen von Indium

Bei dieser Aufgabe dampfen Sie bei verschiedenen Drucken jeweils eine Indium-Schicht durch eine Kreisblende auf eine schwenkbare Plexiglasplatte auf. Es soll jeweils ein Fleck bei einem Druck von 

 * $p\lesssim10^{-4}\,\mathrm{mbar}$ und 
 * $p\lesssim10^{-3}\,\mathrm{mbar}$ 

aufgedampft werden. 

Gehen Sie dabei wie folgt vor: 

 * Evakuieren Sie den RZ mit der TMP; 
 * Schließen Sie V2; 
 * Dampfen Sie den ersten Fleck bei minimal erreichbarem Druck ($p\lesssim10^{-4}\,\mathrm{mbar}$) auf; 
 * Dampfen Sie den zweiten Fleck bei $p\approx10^{-3}\,\mathrm{mbar}$ auf. 

⚠️ **Achtung: Die verfügbare Heizleistung reicht aus, um das VS vollständig zu zerstören!** ⚠️ Der Heizstrom darf daher erst sehr langsam nach oben geregelt werden, nachdem das gewünschte Vakuum für den ersten Fleck erreicht ist. Beobachten Sie das VS beim Hochregeln des Heizstroms. Es soll zwar glühen, aber nicht schmelzen.

 * Belüften Sie die Apparatur.
