### Aufgabe 3.2: Beobachtung höherer Anregungen von Hg

**Hinweise zum Ablauf:**

 * Für diese Aufgabe müssen Sie die FHT anders betreiben, als bisher. Beachten Sie hierzu die Hinweise im Abschnitt **Beobachtung höherer Anregungen von Hg [hier](https://gitlab.kit.edu/kit/etp-lehre/p2-praktikum/students/-/blob/main/Franck_Hertz_Versuch/doc/Hinweise-Versuchsdurchfuehrung.md)**
 * Legen Sie G1 und(!) G2 auf das gleiche Potential $U_{2}$.
 * Senken Sie außerdem ggf. $\vartheta_{S}$ ab.
 * Damit erreichen Sie die folgende Konfiguration:
   * Durch die Reduktion von $\vartheta_{S}$ erhöhen Sie insgesamt die **mittlere freie Weglänge $\lambda$** der Elektronen auf Ihrem Weg durch die FHT.
   * Dadurch, dass $U_{2}$ jetzt bereits an G1 anliegt **verwenden Sie G1 als Anode**.
   * Die Elektronen werden auf der kurzen Strecke zwischen K und G1 auf Ihre volle kinetische Energie beschleunigt.
   * Den Raum zwischen G1 und G2 durchlaufen sie daraufhin, ohne weitere Beschleunigung als **spannungsfreien Driftraum**.
   * Die Strecke zwischen K und G1 ist so kurz, dass die Wahrscheinlichkeit für den Stoß eines Elektrons mit einem Hg-Atom gering ist.
   * Auf dieser Strecke können die Elektronen nun hoch genug beschleunigt werden, um höhere Anregungen bei Stößen mit Hg-Atomen hervorzurufen.  
   * Die Strecke zwischen G1 und G2 ist wiederum lang genug, so dass Stöße der Elektronen mit Hg-Atomen beobachtet werden können.

## Aufgabe 4: Bestimmung der mittleren Energie für die Anregung von Ne durch Elektronenstoß

**Hinweise zum Ablauf:**

 * Für diese Aufgabe benutzen Sie eine **Franck-Hertz-(Ne)-Tetrode** mit indirekter Kathodenheizung, planparalleler Elektrodenanordnung und **Ne-Füllung** (siehe [Datenblatt](https://gitlab.kit.edu/kit/etp-lehre/p2-praktikum/students/-/blob/main/Franck_Hertz_Versuch/Datenblatt.md)).
   * Die Kontaktspannung zwischen der BaO-Kathode und der Metallanode **mindert die angelegte Spannung um $U_{\mathrm{th.}}\approx2.5\,\mathrm{V}$**.
   * Diese Tetrode liegt bereits vollständig verschaltet vor. Diese Schaltung ist der der FHT für Hg ähnlich.
   * Der Verlauf von $I_{A}$ als Funktion von $U_{B}$ kann auf einem bereitstehenden **analogen Oszilloskop** dargestellt und untersucht werden.
   * Ne liegt bei Zimmertemperatur bereits vollständig als Gas vor. Ein **Aufheizen der Tetrode erübrigt sich** daher:
     * Die Teilchenzahldichte $n$ wird beim Heizen nicht erhöht.
     * Die Geschwindigkeit der Ne-Atome nimmt zwar zu, sie ist jedoch nach wie vor vernachlässigbar, im Vergleich zur Geschwindigkeit der Elektronen, nach Durchlaufen der Beschleunigungsstrecke (siehe Abschätzung [hier](https://gitlab.kit.edu/kit/etp-lehre/p2-praktikum/students/-/blob/main/Franck_Hertz_Versuch/doc/Hinweise-mittlere-freie-Weglaenge.md)).
   * Bei den vorherrschenden Anregungen handelt es sich um eine **Gruppe mehrerer Niveaus**, die sich mit dieser Messanordnung nicht getrennt auflösen lassen, Sie können daher nur die **mittlere Anregungsenergie** bestimmen.
   * Die Anregung von Ne ist von rötlich, orangefarbenem, **sichtbarem Licht** begleitet.
   * Wenn Sie die Beschleunigungsspannung **per Hand erhöhen**, sollten Sie beobachten können, wie sich in der Nähe von G2 leuchtende Bänder ausbilden, die sich mit zunehmenden Werten von $U_{2}$ zu G1 hin verlagern. An G2 entstehen mit zunehmender Spannung $U_{2}$ immer neue Bänder. 
 * Die Franck-Hertz-(Ne)-Tetrode ist **nur einmal vorhanden**, Sie müssen sie also mit den anderen Gruppen gemeinsam benutzen.
 * Da die Schaltung bereits aufgebaut ist und Wartezeiten für das Aufheizen oder Abkühlen der Tetrode entfallen, benötigen Sie aber nur wenig Zeit für diese Aufgabe. 
 * **Die Röhre sollte nicht in der für Aufgabe 3.2 modifizierten Schaltung betrieben werden!**


**Hinweise zur Auswertung:**

 * Fügen Sie Ihrer Auswertung eine Photographie der Messanordnung und eine Photographie des **Verlaufs von $U_{A}$ als Funktion von $U_{2}$** zu.
 * Sie sollten auf dem Oszilloskop zwei Maxima ausmachen können aus denen Sie $\Delta U_{B}$ bestimmen können.
 * Schätzen Sie entsprechende Unsicherheiten für die Maxima $U_{B,i}$ ab und pflanzen Sie diese entsprechend auf $\Delta U_{B}$ fort. 

**Anmerkung:**

Die **Ionisierungsenergie von Ne** beträgt $E_{\mathrm{max}}=21.56454(1)\ \mathrm{eV}$ [[1](https://physics.nist.gov/cgi-bin/ASD/ie.pl?spectra=Ne+I&units=1&format=0&order=0&at_num_out=on&sp_name_out=on&ion_charge_out=on&el_name_out=on&seq_out=on&shells_out=on&level_out=on&ion_conf_out=on&e_out=0&unc_out=on&biblio=on&submit=Retrieve+Data)]. Die wichtigsten bekannten **Anregungszustände** finden Sie [hier](https://physics.nist.gov/cgi-bin/ASD/energy1.pl?de=0&spectrum=Ne+I&units=1&format=0&output=0&page_size=15&multiplet_ordered=0&conf_out=on&term_out=on&level_out=on&unc_out=1&j_out=on&lande_out=on&perc_out=on&biblio=on&temp=&submit=Retrieve+Data). Sie liegen zwischen $18.4$ und $19.6\ \mathrm{eV}$. Die Relaxation erfolgt u.a. über zwei Zwischenzustände mit $16.6$ und $16.9\,\mathrm{eV}$, wodurch es zu Energiedifferenzen $\mathcal{O}(2\,\mathrm{eV})$ kommt, die zur Abstrahlung von sichtbarem Licht im rötlich, orangefabenen Farbbereich führen.
