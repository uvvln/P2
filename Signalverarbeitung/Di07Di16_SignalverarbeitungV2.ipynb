### Aufgabe 2.2: Verstärkungsgesetz

 * Ersetzen Sie einen der Widerstände durch ein Potentiometer und messen Sie für fünf verschiedene Konfigurationen des Potentiometers den Verstärkungsfaktor.
 * Überprüfen Sie den funktionalen Zusammenhang für den Verstärkungsfaktor $(1 + \frac{R_1}{R_2})$ durch eine entsprechende Modellanpassung.

</div>

**V E R S U C H S B E S C H R E I B U N G**

In diesem Versuch wird das Verstärkungsgesetz des nicht-invertierenden Operationsverstärkers experimentell überprüft. Dazu wird die Schaltung aus 2.1 erweitert, indem der Widerstand $R_1$ durch ein Potentiometer ersetzt wird. Dadurch kann man das Modell

\begin{equation*}
U_{\mathrm{out}} = U_{\mathrm{in}} \left(1 + \frac{R_1}{R_2}\right)
\end{equation*}

näher untersuchen.

Die Spannungsverstärkung $\beta$ ist:
\begin{equation*}
\beta = 1 + \frac{R_1}{R_2}
\end{equation*}

Für ein sinusförmiges Eingangssignal ($\nu \approx 1\,\mathrm{kHz}$, Amplitude $2.8 \pm 0.2\,\mathrm{V}$) werden sechs verschiedene Verstärkungen im Bereich $U_{\mathrm{out}} = 3 \ldots 9\,\mathrm{V}$ eingestellt. Für jede Einstellung wird das Potentiometer angepasst, um andere Verstärungsfaktoren zu erhalten. $R_2 = 1.15 \, k\Omega $ bleibt dabei fest. Für beide Widerstände wird von einem relativen Fehler von 1 % ausgegangen. Auf $U_{\mathrm{in}}$ und $U_{\mathrm{out}}$ wird eine absolute Unsicherheit von 0.2 V angenommen.

Ein- und Ausgangsspannung werden mit dem Oszilloskop gemessen. Aus den Messwerten werden die Größen $(U_{\mathrm{out}}/U_{\mathrm{in}},\, R_1/R_2)$ bestimmt und zur Überprüfung des Modells gefittet. 

---


Auch hier werden Daten von anderen Gruppen verwendet:

| $R_1$ / $\mathrm{k} \Omega$ | $U_{\mathrm{out}}$ / V |
|------------------|-----------------------|
| 0.22             | 3.0                   |
| 0.53             | 4.4                   |
| 1.01             | 5.2                   |
| 1.48             | 7.2                   |
| 1.83             | 7.4                   |
| 2.10             | 8.8                   |

Wenn man diese Daten auf das oben aufgeführte Modell fittet, erhält man:

```python
from kafe2 import config, XYContainer, Fit, XYFit,  Plot
import numpy as np
from uncertainties import ufloat
import uncertainties.unumpy as unp
import matplotlib.pyplot as plt

# Widerstände in kOmega
R1 = .220
R2 = .530
R3 = 1.010
R4 = 1.480
R5 = 1.830
R6 = 2.100

R_1= np.array([R1, R2, R3, R4, R5, R6])
un_rel_R = 0.01
uR_1 = unp.uarray(R_1, un_rel_R * R_1)

# Ausgangsspannungen in V
U_1 = 3
U_2 = 4.4
U_3 = 5.2
U_4 = 7.2
U_5 = 7.4
U_6 = 8.8

U_out = np.array([U_1, U_2, U_3, U_4, U_5, U_6])
un_abs_U = 0.2
uU_out = unp.uarray(U_out, un_abs_U)

# Eingangsspannung in V
U_in = 2.8
uU_in = ufloat(U_in, un_abs_U)

# anderer Widerstand in kOmega:
R_2 = 1.15
uR_2 = ufloat(R_2, un_rel_R * R_2)

# Model fit

def amplification(R_1, R_2, U_in): 
    return U_in * (1 + R_1 / R_2)


xy_data = XYContainer(x_data=unp.nominal_values(uR_1), # R_2 werte als x-Achse
                      y_data=unp.nominal_values(uU_out))  # U_out werte als y-Achse

xy_data.add_error(
    axis='y', 
    err_val=unp.std_devs(uU_out)
)  
xy_data.add_error(
    axis='x', 
    err_val=unp.std_devs(uR_1)
) 

fit = Fit(data=xy_data, model_function=amplification)

# constrain R_2 and U_in
fit.add_parameter_constraint(name='R_2',   value=uR_2.nominal_value,   uncertainty=uR_2.std_dev)
fit.add_parameter_constraint(name='U_in',   value=uU_in.nominal_value,   uncertainty=uU_in.std_dev)

fit.do_fit()

# plot the fit
plot = Plot(fit_objects=fit)

plot.x_label = '$R_1$ in k$\Omega$' 
plot.y_label = '$U_{out}$ in V'

plot.plot()
plt.show()
```

**D I S K U S S I O N**

Vom dem Verlauf der Datenpunkte in der Graphik kann man sehen, dass die Annahme, dass $U_{\mathrm{out}}$ in $R_1$ linear ist, sehr wahrscheinlich korrekt ist. Allerdings ist der $\chi^2/ \alpha$ - Wert mit 3.562 etwas größer als 1. Das muss nicht auf ein unzureichendes Modell hinweisen. Da der Verlauf der Datenpunkte eher für dieses Modell sprechen, kann man davon ausgehen, dass die Unsicherheiten der Messung zu wenig konservativ abgeschätzt wurden. Da $R_1$ mit einem Multimeter sehr genau bestimmt werden konnte, wird davon ausgegangen, dass das Problem in der Abschätzung der Unsicherheit der Spannung $\Delta U = 0.2 V$ liegt. Diese sollte höher angenommen werden.

Das Eingangssignal $U_{\mathrm{in}}$ wird über den Widerstand $R_1$ in den invertierenden Eingang eingespeist, während sich im Rückkopplungszweig der Widerstand $R_2$ befindet. Der nicht-invertierende Eingang ist mit Masse verbunden, sodass am invertierenden Eingang aufgrund der goldenen Regeln ebenfalls virtuelle Masse anliegt.

Da kein Strom in den OPV fließt, ist der Strom durch beide Widerstände identisch. Mit dem ohmschen Gesetz ergibt sich:
\begin{equation*}
U_{\mathrm{in}} = R_1 I, \quad U_{\mathrm{out}} = -R_2 I
\end{equation*}
Daraus folgt für die Ausgangsspannung:
\begin{equation*}
U_{\mathrm{out}} = -U_{\mathrm{in}} \frac{R_2}{R_1}
\end{equation*}

Die Schaltung bewirkt somit eine Invertierung (Phasenverschiebung um $\pi$) und eine Verstärkung mit dem Faktor $\beta = \frac{R_2}{R_1}$.

Für den Versuch werden Widerstände mit $R_1 = 1.15\,\mathrm{k\Omega}$ und $R_2 = 2.26 \,\mathrm{k\Omega}$ verwendet, sodass durch Tauschen der Widerstände zwei verschiedene Verstärkungsfaktoren realisiert werden können. Als $U_{\mathrm{in}}$ wird eine Sinusspannung ($\nu \approx 1\,\mathrm{kHz}$, Amplitude $3\,\mathrm{V}$) eingespeist. 

Es werden 2 unterschiedliche Verstärungsfaktoren $\beta$ erwartet:

\begin{equation*}
\beta_1 = \frac{R_1}{R_2} \approx 0.5 , \quad \beta_2 = \frac{R_2}{R_1} \approx 2
\end{equation*}

Zur Überprüfung werden Ein- und Ausgangsspannung mit dem Oszilloskop dargestellt und die Verstärkung für beide Widerstandskonfigurationen qualitativ bestimmt.

In diesem Versuch wird ein invertierender Summierverstärker aufgebaut (s. Skizze) und untersucht. Zunächst wird die Wirkung einer einzelnen Eingangsspannung $U_1$ betrachtet. Diese wird über den Widerstand $R_1$ eingespeist, wobei der gesamte Strom über den Rückkopplungswiderstand $R_f$ abfließt. Es gilt:
\begin{equation*}
\frac{U_1}{R_1} = -\frac{U_{\mathrm{out}}}{R_f}
\quad \Rightarrow \quad
U_{\mathrm{out}} = -U_1 \frac{R_f}{R_1}
\end{equation*}

Werden zusätzlich mehrere Eingangsspannungen berücksichtigt, so addieren sich die Ströme am invertierenden Eingang. Für zwei Eingangsspannungen $U_1$ und $U_2$ folgt:
\begin{equation*}
\frac{U_1}{R_1} + \frac{U_2}{R_2} = -\frac{U_{\mathrm{out}}}{R_f}
\quad \Rightarrow \quad
U_{\mathrm{out}} = -\left( U_1 \frac{R_f}{R_1} + U_2 \frac{R_f}{R_2} \right)
\end{equation*}

Für den Versuch werden die Widerstände $R_1 = 6{,}81\,\mathrm{k\Omega}$, $R_2 = 9{,}09\,\mathrm{k\Omega}$ und $R_f = 1{,}15\,\mathrm{k\Omega}$ verwendet. Die Eingangs- sowie Ausgangsspannung wird mit dem Oszilloskop dargestellt.

 Nach der oben dargestellten Formel sollte sich das Ausgangssingal zu

\begin{equation*}
U_{\mathrm{out}} \approx - 3 \, V
\end{equation*}

transformieren. Also ein invertierte Ausgangsspannung von $\approx 3 \, V$ ist zu erwarten.

In diesem Versuch wird die in Aufgabe 3.2 aufgebaute Summierverstärker-Schaltung um einen zusätzlichen Inverter ergänzt. Der Inverter hat eine Verstärkung von $\beta = -1$ (d.h. $R_1 = R_f$) und wird am Ausgang der bestehenden Schaltung angeschlossen.

Der Summierverstärker aus Aufgabe 3.2 liefert eine invertierte, gewichtete Summe der Eingangsspannungen:
\begin{equation*}
V_{\mathrm{out'}} = -\left( V_1 \frac{R_f}{R_1} + V_2 \frac{R_f}{R_2} \right)
\end{equation*}

Durch den nachgeschalteten Inverter ergibt sich insgesamt:
\begin{equation*}
V_{\mathrm{out}} = - V_{\mathrm{out'}} = 
\left( V_1 \frac{R_f}{R_1} + V_2 \frac{R_f}{R_2} \right)
\end{equation*}

Die doppelte Invertierung hebt die Phasendrehung auf, sodass das Ausgangssignal nicht mehr invertiert ist. Gleichzeitig bleibt die gewichtete Skalierung der Eingangsspannungen erhalten. Die Schaltung wirkt damit ähnlich wie ein Spannungsteiler, jedoch aktiv und mit einstellbaren Verstärkungsfaktoren.

Zur Überprüfung wird eine Sinusspannung ($\nu \approx 1\,\mathrm{kHz}$, Amplitude $9\,\mathrm{V}$) angelegt und Ein- sowie Ausgangsspannung mit dem Oszilloskop dargestellt. Im Unterschied zum passiven Spannungsteiler erfolgt die Spannungsanpassung hier aktiv über Operationsverstärker. Der OPV ist an einer Versorgungsspannung angeschlossen, was bedeutet, dass Leistung in das System eingespeist werden kann. Im Fall des passiven Spannungsteilers ist die Ausgangsspannung somit auf die Eingangsspannung begrenzt. Beim OPV ist der limitierende Faktor die Versorgungsspannung.
