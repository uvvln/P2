</div>

 * Legen Sie zur Überprüfung der Schaltung am Funktionsgeber (FG) eine **Sinusspannung mit mittlerer Frequenz** (z.B. $\nu=1\,\mathrm{kHz}$) und einer Amplitude von $9\,\mathrm{V}$ für $U_{\mathrm{in}}$ an und stellen Sie **sowohl $U_{\mathrm{in}}$ als auch $U_{\mathrm{out}}$ auf dem Oszilloskop** geeignet dar.

    * Beschreiben Sie, wie die Verzerrung des Rechtecksignals zustande kommt.
