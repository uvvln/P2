## Aufgabe 1: Messung des absoluten Nullpunkts der Temperatur

</div>

🦉 **Für diese Aufgabe benötigen Sie die Grundlagen aus den folgenden Dateien:** 🦉

 * [Grundbegriffe der Thermodynamik](https://gitlab.kit.edu/kit/etp-lehre/p2-praktikum/students/-/blob/main/Ideales_und_reales_Gas/doc/Hinweise-Thermodynamik.md)
 * [Messung des absoluten Nullpunkts der Temperatur mit Hilfe eines Gasthermometers](https://gitlab.kit.edu/kit/etp-lehre/p2-praktikum/students/-/blob/main/Ideales_und_reales_Gas/doc/Hinweise-Gasthermometer.md)

🔧 **Hinweise zum Ablauf:** 🔧 

 * ⚠️ Die Annahme $\vartheta=const.$ bei der Side- und Schmelztemperatur setzt ein **thermodynamisches Gleichgewicht** voraus, das insbesondere bei Eiswasser nicht immer gegeben ist.
    * Schieben Sie das Becherglas vorsichtig von unten kommend, so hoch wie möglich über K.
    * Füllen Sie es dann mit einem Gemisch aus möglichst _fein_ zerstoßenem Eis und kaltem Wasser, so dass K möglichst ganz bedeckt ist.
    * Rühren Sie die Mischung vorsichtig durch, so dass eine möglichst homogene Temperaturverteilung im Bereich $\vartheta=0^{\circ}\mathrm{C}$ im Wärmebad entsteht. 
    * Die Luft in K beginnt daraufhin abzukühlen, worauf in K ein Unterdruck relativ zur Umgegung entsteht.
       * ⚠️ Justieren Sie, während dieses Vorgangs RR **bis zum Anschlag nach unten um zu verhindern, dass Hg in K eingesogen wird**. 
    * Nach ${\approx}15\ \mathrm{min}$ sollte die Luft in K die Temperatur des Wärmebads angenommen haben.
       * Regeln Sie die Höhe der Hg-Kuppe in RL so ein, dass sie gerade die Spitze S berührt.
       * Sie erreichen dies, indem Sie RR langsam nach oben verschieben.
       * Aus der Höhendifferenz $\Delta h$ der beiden Quecksilberkuppen können Sie $p_{0}$ bestimmen.
       * Um den Druck in $\mathrm{mbar}$ zu erhalten verwenden Sie die Umrechnung:

$$
1\,\mathrm{mm\,Hg} = 1.33322\,\mathrm{mbar}
$$

 * Bringen Sie dannn das Wasser in W auf Siedetemperatur $\vartheta_{s}$.
 * Beachten Sie dabei die folgenden Punkte: 
    * $\vartheta_{s}$ ist selbst druckabhängig. Eine **Dampfdruckkurve für Wasser**, ebenso wie ein digitales Barometer zur Bestimmung des Umgebungsdrucks liegen in den Versuchsräumen aus.
    * ⚠️ Achten Sie auch beim Heizen darauf, dass K immer **vollständig mit Wasser bedeckt** ist. 
    * ⚠️ Bevor Sie $p(\vartheta_{s})$ wie oben bestimmen, sollte das Wasser einige Minuten lang sieden, um sicher zu stellen, dass sich W im **thermischen Gleichgewicht** befindet.

---

🖥️ **Hinweise zur Auswertung:** 🖥️
 
 * ✍🏻 **Protokollieren Sie:** ✍🏻 
    * Beschreiben Sie Ihr Vorgehen.
    * Wenn Sie $\gamma$ genauer bestimmen wollen, sollten Sie sich **zu einigen zusätzlichen Effekten** Gedanken machen, die Ihre Messung verfälschen können:
       * K hat selbst eine **thermische Ausdehnung**;
       * K verformt sich unter Druck;
       * Luft ist kein ideales Gas;
       * Ein Teil des Gases befindet sich (bei Raumtemperatur) in RZ, außerhalb von K.
    * Alle Effekte bis auf die thermische Ausdehnung von K sind tatsächlich sehr klein, so dass Sie sie in Ihrer Auswertung vernachlässigen können. Sie sollten Sie aber dennoch diskutieren.
    * Geben Sie in einer Tabelle an, ob eine entsprechende Variation $\gamma$ vergößert oder verkleinert.
    * Die thermische Ausdehnung von K sollten Sie als **Korrektur in Ihrer Auswertung berücksichtigen**.
    * Geben Sie $\vartheta_{0}\pm\vartheta_{0}$ und die Größe der Korrektur $\delta\vartheta$ an.

### Aufgabe 2.1: Messung nach der Methode von [Clément-Desormes](https://de.wikipedia.org/wiki/Experiment_von_Cl%C3%A9ment-Desormes)

</div>

🔧 **Hinweise zum Ablauf:** 🔧 

 * Beschreiben Sie zur Vorbereitung auf den Versuch den Prozess zur Bestimmung von $\kappa$ in eigenen Worten. 
 * ⚠️ Erzeugen Sie mit dem Handblasebalg **nicht zu viel Überdruck $\Delta p_{1}$**, da sonst beim Druckausgleich im Übergang von **(1) nach (2) in Abbildung 1 (b) [hier](https://gitlab.kit.edu/kit/etp-lehre/p2-praktikum/students/-/blob/main/Ideales_und_reales_Gas/doc/Hinweise-Clement-Desormes.md)** Flüssigkeit vom Flüssigkeitsmanometer in K eindringen kann. 
 * 💡 Achten Sie darauf, dass sich ein **Gleichgewichtszustand** eingestellt hat bevor Sie das Manometer ablesen.  
 * 💡 Ein **Schwachpunkt dieser Messung** besteht in der Trägheit der Manometerflüssigkeit, die beim Übergang von **(2) nach (3)  in Abbildung 1 (b) [hier](https://gitlab.kit.edu/kit/etp-lehre/p2-praktikum/students/-/blob/main/Ideales_und_reales_Gas/doc/Hinweise-Clement-Desormes.md)** in Schwingung versetzt wird. Diese beeinflusst den Druck beim Schließen des Ventils und damit direkt die Bestimmung von $\Delta p_{2}$.
    * Versuchen Sie mit dem Schließen des Ventils den **Nulldurchgang der Schwingung abzupassen**. Dies wird einfacher, wenn Sie so weit wie möglich das Abklingen der Schwingung abwarten, wobei trotzdem $\delta Q\approx 0$ erfüllt bleiben sollte.
    * Wir geben eine Wartezeit von ${\approx}3\ \mathrm{s}$ als Richtwert vor.  
 * Führen Sie eine Messreihe mit mindestens **10 Einzelmessungen** durch und bestimmen Sie Messwert und Unsicherheit für $\kappa$ aus Mittelwert und Varianz dieser Stichprobe.
 * Führen Sie ein Messreihe bei **etwas längerer und kürzerer Öffnungszeit des Ventils** durch, um ein Gefühl für den Einfluss der Öffnungszeit auf die Messung zu bekommen.

🖥️ **Hinweise zur Auswertung:** 🖥️

    * Beschreiben Sie Ihr Vorgehen.

### Aufgabe 2.2: Messung nach der Methode von [Rüchardt](https://de.wikipedia.org/wiki/R%C3%BCchardt-Experiment)

</div>

🦉 **Für diese Aufgabe benötigen Sie die Grundlagen aus den folgenden Dateien:** 🦉

 * [Messung des Adiabatenexponent nach der Methode von Rüchardt](https://gitlab.kit.edu/kit/etp-lehre/p2-praktikum/students/-/blob/main/Ideales_und_reales_Gas/doc/Hinweise-Ruechardt.md)

🔧 **Hinweise zum Ablauf:** 🔧 

 * Beschreiben Sie zur Vorbereitung auf den Versuch den Prozess zur Bestimmung von $\kappa$ in eigenen Worten.
 * ⚠️ **Sie haben dann die Möglichkeit entweder die Originale Messung nach Rüchardt oder einen Versuch mit elektronischer Auslese durchzuführen.** ⚠️

---

**Originalmethode nach Rüchardt:**

---

**Diese Messung führen Sie nur für Luft durch.**  

 * In diesem Fall erzeugen Sie die Schwingung mit Hilfe einer **Stahlkugel als Pfropfen**, deren Durchmesser exakt mit dem Innendurchmesser des Glasrohrs übereinstimmt, so dass sie das Glasrohr nahezu luftdicht verschließt.
 * Das Glasrohr wird mit Hilfe eines durchbohrten Stopfens ebenfalls möglichst luftdicht auf eine der bereitgestellten $10\ \mathrm{l}$-Flaschen aufgesetzt. 
 * 💡 Bei dieser Anordnung **schwingt die Kugel auf einem Luftpolster mit großem Volumen $V$ während $A$ klein ist, $T_{\Omega}$ ist daher groß genug**, so dass Sie die Schwingung gut beobachten und $T_{\Omega}$ gut mit einer Stoppuhr bestimmen können.
 * 💡 Nehmen Sie für das Volumen dieses Luftkissens $V=10.58\ \mathrm{l}\pm0.3\%$ an.
 * ⚠️ Bei Glasrohr und Stahlkugel handelt es sich um **Präzisionsanfertigungen, die mit größter Sorgfalt zu behandeln sind!** Damit die Kugel das Glasrohr luftdicht abschließen, sich aber gleichzeitig möglichst reibungsfrei darin bewegen kann müssen die folgenden Bedingungen so gut wie möglich erfüllt sein: 
    * Das Glasrohr muss möglichst genau senkrecht stehen.
    * Sowohl das Glasrohr, als auch die Stahlkugel müssen möglichst sauber sein. 
    * Alle Stopfen müssen luftdicht abschließen.
 * Reinigen Sie Kugel und Rohrinnenfläche sorgfältig mit einem bereitliegenden Lederlappen. ⚠️ **Berühren Sie die Kugel nach Möglichkeit niemals mit bloßen Fingern.** Sollte dies dennoch geschehen, wiederholen Sie den Reinigungsvorgang.
 * Setzen Sie das Glasrohr so ein, dass der durchbohrte Stopfen möglichst luftdicht abschließt. 
 * Lassen Sie die Kugel aus dem Lederlappen behutsam ins Glasrohr gleiten und bestimmen Sie $T_{\Omega}$ aus möglichst vielen, **mindestens aber 5 Schwingungen**. 
 * 💡 **Wiederholen Sie diesen Vorgang mehrfach, mindestens aber 10 mal**, um ein Maß für die Streuung zu erhalten. 
 * Um die Kugel nach Beendigung einer Messreihe aus dem Rohr zu entnehmen, kippen Sie die Flasche –so lange die Kugel sich noch in der Glasröhre befindet– vorsichtig um und lassen Sie die Kugel in die bereitstehende Plastikschale gleiten. Beachten Sie die Hinweise [hier](https://gitlab.kit.edu/kit/etp-lehre/p2-praktikum/students/-/blob/main/Ideales_und_reales_Gas/doc/Hinweise-Ruechardt.md).

---

**Methode mit dem Kolbenprober:**

---

**Diese Messung führen Sie für Luft und für das Edelgas Ar durch.**  

 * In diesem Fall **ersetzt der Kolbenprober das Glasrohr und die Kugel**; 💡 $V$ und somit $T_{\Omega}$ sind deutlich kleiner, als bei der originalen Anordnung von Rüchardt, so dass $T_{\Omega}$ nicht mehr mit dem Auge sondern elektronisch erfolgt!
 * Wir tun dies mit Hilfe des folgenden Aufbaus:
    * Am Kolben befindet sich ein Magnet.
    * Um die Fassung des Kolbens ist eine Induktionsspule gewickelt.
    * Die Induktionsströme werden von einem Frequenzzähler aufgenommen. 
 * 💡 Indem Sie die **Induktionsspule entlang des Pfropfens verschieben**, können Sie die Werte von $V$ selbst bestimmen.
    * Messen Sie **mindestens 5 Werte** für $T_{\Omega}$ zwischen $V=30-80\ \mathrm{ml}$. 
 * Führen Sie diese Messung **erst mit Luft und dann mit Ar** durch.

---

🖥️ **Hinweise zur Auswertung:** 🖥️
 
 * ✍🏻 **Protokollieren Sie:** ✍🏻 
    * Beschreiben Sie Ihr Vorgehen.
    * Bestimmen Sie $\kappa\pm\Delta\kappa$.
    * Bei der **Originalmethode nach Rüchardt** bestimmen Sie $\kappa\pm\Delta\kappa$ aus einer Messreihe von mindestens 10 Einzelmessungen.
    * Bei der **Methode mit de Kolbenprober** liegen Ihnen fünf Messpunkte $(V_{i},T_{\Omega,i})$ vor.
       * Aus diesen Messpunkten können Sie $\kappa\pm\Delta\kappa$ durch Anpassung eines geeigneten Modells $T_{\Omega}(V)$ nach Gleichung **(3) [hier](https://gitlab.kit.edu/kit/etp-lehre/p2-praktikum/students/-/blob/main/Ideales_und_reales_Gas/doc/Hinweise-Ruechardt.md)** bestimmen.
       * Da Sie zudem die Messung nicht nur für Luft, sondern auch auf Ar durchführen können Sie überprüfen Sie, ob $\kappa$ mit der Erwartung $\kappa = \frac{f+2}{f}$ aus der [kinetischen Gastheorie](https://de.wikipedia.org/wiki/Kinetische_Gastheorie) übereinstimmt, wobei $f$ der Anzahl der [Freiheitsgrade](https://de.wikipedia.org/wiki/Freiheitsgrad) des untersuchten Gases entspricht.
    * Je nach Gas erwarten Sie die folgenden Werte für $f$:
       * Edelgas, z.B. Ar: $f=3$,
       * $\mathrm{O_{2}},\ \mathrm{N_{2}}: f=5$,
       * $\mathrm{CO_{2}}: f=7$.
    * Diskutieren Sie die Übereinstimmung von $\kappa\pm\Delta\kappa$ für Luft aus **Aufgabe 2.1** und dieser Aufgabe.
    * Beurteilen sie, welches die genauere Messung ist.

## Aufgabe 3: Dampfdruckkurve von n-Hexan

</div>

🦉 **Für diese Aufgabe benötigen Sie die Grundlagen aus den folgenden Dateien:** 🦉

 * [Phasendiagramm und Dampfdruckkurve](https://gitlab.kit.edu/kit/etp-lehre/p2-praktikum/students/-/blob/main/Ideales_und_reales_Gas/doc/Hinweise-Dampfdruckkurve.md)

🔧 **Hinweise zum Ablauf:** 🔧 

 * Bei diesem Versuch beobachten Sie den **Phasenübergang zwischen flüssig und gasförmig** von n-Hexan. 
 * In einem Glaskolben K befinden sich nur die Flüssigkeit und der Dampf von n-Hexan. Der Dampfdruck wird mit einem direkt verbundenen Hg-Manometer gemessen, dessen Stand Sie mit einem [Kathetometer](https://de.wikipedia.org/wiki/Kathetometer) ablesen können.
 * ⚠️ Tauchen Sie den Kolben mit dem n-Hexan, so vollständig wie möglich, in ein Wärmebad, dessen Temperatur Sie **so langsam(!) verändern müssen, dass die Flüssigkeit und der Dampf an jedem Messpunkt im Gleichgewicht sind**. Der Erfolg Ihrer Messung hängt davon ab, wie gut es Ihnen gelingt das n-Hexan während des Vorgangs im thermischen Gleichgewicht zu halten.
 * Stellen Sie das Fernrohrokular des Kathetometers auf das Fadenkreuz ein. ⚠️ **Während der Messung darf am Fernrohr selbst daraufhin nichts mehr verändert werden.** 
    * Zur Messung des Dampfdrucks bei Raumtemperatur ($T_{0}$) visieren Sie die beiden Hg-Kuppen nacheinander an und lesen die Höhendifferenz $h_{0}$ an der Kathetometerskala ab. ⚠️ **Den Fokus auf die Kuppen müssen Sie dabei durch Verschieben des ganzen Gestells erreichen.**
    * Für die weiteren Druckmessungen genügt es, wenn Sie nur noch eine der beiden Kuppen anvisieren.
    * 🔔 Achten Sie darauf, dass zu Beginn des Experiments **kein flüssiges n-Hexan außerhalb des Wärmebads** niedergeschlagen ist, was Ihre Messung verfälschen würde.
    * Zu Behandlung etwaiger Fälle dieser Art beachten Sie die Hinweise [hier](https://gitlab.kit.edu/kit/etp-lehre/p2-praktikum/students/-/blob/main/Ideales_und_reales_Gas/doc/Hinweise-Dampfdruckkurve.md).
    * Falls Ihre Apparatur zu Beginn der Messung noch Niederschlag von n-Hexan auf dem Hg-Barometer aufweisen sollte vermerken Sie diesen Umstand bitte genau in Ihrem Protokoll. 
 * Rühren Sie während des Versuchs das Wärmebad langsam um, so dass Sie eine **möglichst homogene Temperaturverteilung** erhalten.
 * Nehmen Sie zunächst die Dampfdruckkurve bei **langsam sinkender Temperatur** auf. Geben Sie hierzu nach und nach möglichst fein zerstoßene Eisstückchen ins Wärmebad.
 * Nehmen Sie dann die Dampfdruckkurve bei **langsam steigender Temperatur** auf. Gießen Sie hierzu dem Wärmebad destilliertes Wasser mit Raumtemperatur zu.
 * Sie erhalten hieraus zwei Messreihen.

---

🖥️ **Hinweise zur Auswertung:** 🖥️
 
 * ✍🏻 **Protokollieren Sie:** ✍🏻 
    * Beschreiben Sie Ihr Vorgehen.
    * Geben Sie den von Ihnen bestimmten Wert für $h_{0}\pm\Delta h_{0}$ in Ihrer Auswertung an.
    * Beurteilen Sie die **Gleichheit der Messpunkte** aus beiden Messreihen bei jeweils gleicher Temperatur.
    * Die größte Schwierigkeit bei dieser Messung besteht darin:
       * das n-Hexan zu jedem Zeitpunkt im Gleichgewicht zu halten,
       * beim Abkühlen den Dampf nicht zu übersättigen und
       * beim anschließenden Erwärmen in der Flüssigkeit keinen Siedeverzug zu provozieren.
 * Je besser Sie beim Messen mit diesen Schwierigkeiten zurechtgekommen sind desto besser stimmen die Datenpunkte Ihrer Messreihen überein.
 * Schätzen Sie durch **Anpassung eines geeigneten Modells**, z.B. nach Gleichung **(5) [hier](https://gitlab.kit.edu/kit/etp-lehre/p2-praktikum/students/-/blob/main/Ideales_und_reales_Gas/doc/Hinweise-Dampfdruckkurve.md)** die molare Verdampfungswärme $Q_{\mathrm{M}}\pm\Delta Q_{\mathrm{M}}$ von n-Hexan ab.
 * Stellen Sie Ihre Messpunkte $(T_{i},\Delta h_{i})$ dazu geeignet dar.
 * Geben Sie den $\chi^{2}$-Wert der Anpassung normiert auf die Freiheitsgrade des Modells an.
 * Beantworten Sie die Frage, welchen Einfluss der Dampfdruck von Hg auf die Messung des Verlaufs der Dampfdruckkurve des n-Hexan hat.
