## Aufgabe 1: Messung des absoluten Nullpunkts der Temperatur

 * Bestimmen Sie den [**Volumenausdehnungskoeffizienten**](https://de.wikipedia.org/wiki/Ausdehnungskoeffizient) $\gamma$ von Luft mit Hilfe eines Gasthermometers.
 * Berechnen Sie daraus die **Temperatur des absoluten Nullpunkts** $\vartheta_{0}$ in $^{\circ}\mathrm{C}$.
 * Überprüfen Sie den Effekt, den die **Ausdehnung des Glaskolbens K** als Funktion von $\vartheta$ hat und korrigieren Sie Ihre Messung ggf. entsprechend. 
 
</div>

**V E R S U C H S B E S C H R E I B U N G**

Ziel dieses Experiments ist die Bestimmung des Volumenausdehnungskoeffizienten $\gamma$ von Luft mithilfe eines Gasthermometers. Daraus wird anschließend die Temperatur des absoluten Nullpunkts $\vartheta_0$ in $^{\circ}\mathrm{C}$ berechnet. Zusätzlich wird der Einfluss der thermischen Ausdehnung des Glaskolbens berücksichtigt und gegebenenfalls korrigiert.

Das Gasthermometer basiert auf einem eingeschlossenen Gasvolumen $V = \mathrm{const.}$ und einer konstanten Stoffmenge $n = \mathrm{const.}$. Aus der idealen Gasgleichung ergibt sich

$$
p(V,T) = \frac{nRT}{V}
$$

Für konstantes Volumen gilt das Gesetz von Amontons:

$$
p \propto T
$$

In Celsius-Temperatur geschrieben folgt daraus

$$
p(\vartheta)\big|_{V=\mathrm{const.}} = p_0 \left(1 + \gamma \, \vartheta \right)
$$

mit $p_0 = p(\vartheta = 0^\circ \mathrm{C})$. Der Volumenausdehnungskoeffizient $\gamma$ beschreibt dabei die lineare Änderung des Drucks mit der Temperatur.

Der absolute Nullpunkt ergibt sich aus der Extrapolation der linearen Beziehung zu dem Punkt, an dem der Druck verschwindet:

$$
\vartheta_0 = -\frac{1}{\gamma}
$$

Der Druck im Gasvolumen wird über die Höhendifferenz $\Delta h$ der Quecksilbersäulen bestimmt:

$$
p(\Delta h) = p_{\mathrm{norm}} + \rho(\mathrm{Hg}) \, g \, \Delta h
$$

Dabei ist $p_{\mathrm{norm}}$ der Umgebungsdruck, $\rho(\mathrm{Hg})$ die Dichte von Quecksilber und $g$ die Erdbeschleunigung. Zur Umrechnung gilt

$$
1\,\mathrm{mmHg} = 1.33322\,\mathrm{mbar}
$$

Zur Bestimmung von $\gamma$ werden Messungen nahe der Schmelztemperatur $\vartheta \approx 0^\circ \mathrm{C}$ sowie bei der Siedetemperatur $\vartheta_s$ von Wasser durchgeführt. Daraus ergibt sich

$$
\gamma = \frac{p(\vartheta_s) - p_0}{p_0 \, \vartheta_s}
$$

Für die Messung wird der Glaskolben vollständig in ein Eis-Wasser-Gemisch getaucht, sodass eine möglichst homogene Temperatur bei $\vartheta = 0^\circ \mathrm{C}$ entsteht. Das System muss ausreichend Zeit haben, um ein thermisches Gleichgewicht zu erreichen. Das ist auch der Grund, warum Schmelz- und Siedetemperatur zur Datenentnahme ausgewählt wurden, da es einfach ist, diese Temperatur zu halten. Die Quecksilbersäule wird so eingestellt, dass sich ein stabiler Gleichgewichtszustand einstellt, aus dem $p_0$ bestimmt werden kann. Beim Erwärmen auf Siedetemperatur muss ebenfalls sichergestellt werden, dass das Wasser vollständig siedet und der Kolben konstant auf Temperatur gehalten wird, bevor die Messung erfolgt.

Die Temperaturabhängigkeit des Siedepunkts ist druckabhängig, daher muss der Umgebungsdruck berücksichtigt werden. Die Druckdifferenz wird erneut über $\Delta h$ bestimmt. Die gesamte Methode beruht auf der Annahme eines linearen Zusammenhangs zwischen Druck und Temperatur bei konstantem Volumen, wodurch eine Extrapolation bis zum Schnittpunkt mit der Temperaturachse möglich ist und damit der absolute Nullpunkt bestimmt wird.

---

```python
import numpy as np
from uncertainties import ufloat

p_norm = ufloat(101325, 200) # in Pa
umrechnung_hg = ufloat(1.333322, 0.000001)
uT_0 = ufloat(5.7, 0.1) #in C
uT_s = ufloat(99.9, 0.1)
udelta_h_0 = ufloat(-3.5, 0.2)*10# in mm
udelta_h_s = ufloat(23.1, 0.2)*10# in mm

p0 = p_norm + umrechnung_hg * udelta_h_0 * 100  # mbar in Pa
ps = p_norm + umrechnung_hg * udelta_h_s * 100

ugamma = (ps - p0) / (uT_s * p0)

absolute_zero = -1 / ugamma

print("gamma =", ugamma)
print("T_0 =", absolute_zero)
```

Mit diesem Vorgehen erhält man einen Wert für den absoluten Nullpunkt ($\vartheta_0 = -272.3 \pm 3.5$ °C), der sehr nahe am Referenzwert ($\vartheta_0^{\text{ref}} = -273.15$ °C). Neben der idealisierten Beschreibung können systematische Effekte auftreten. Dazu zählen die thermische Ausdehnung des Glases, eine mögliche elastische Verformung von $K$, Nichtidealität der Luft sowie Gasanteile im Rohrsystem $RZ$.

Die thermische Ausdehnung von $K$ ist dabei der wichtigste systematische Beitrag, da die anderen Effekte relativ klein sind. Mit

$$
\gamma_{\mathrm{K}} = (0.9 \pm 0.1)\times 10^{-5}\,\mathrm{K}^{-1}
$$

ändert sich das Volumen zu

$$
\frac{V_{\mathrm{K}}'}{V_{\mathrm{K}}} = 1 + \vartheta_s \gamma_{\mathrm{K}}.
$$

Dadurch wird der gemessene Druck bei $\vartheta_s$ geringfügig reduziert, was eine Korrektur von $\gamma$ erfordert:

$$
\gamma^{(1)} = \frac{(1 + T_s \gamma_{\mathrm{K}})\,p(\vartheta_s) - p_0}{p_0\,\vartheta_s}
$$

mit Zerlegung in

$$
\gamma^{(1)} = \gamma^{(0)} + \delta\gamma,
\quad
\gamma^{(0)} = \frac{p(\vartheta_s) - p_0}{p_0\,\vartheta_s}
$$

wobei 

$$
T_s = \vartheta_s - \frac{1}{\gamma}.
$$

```python
# berechnung mit korrejtur
T_s = uT_s - 1 / ugamma
ugamma_K = ufloat(0.9, 0.1)*10**(-5) 
ugamma_1 = ((1 + T_s * ugamma_K) * ps - p0) / (uT_s * p0)

absolute_zero_1 = -1 / ugamma_1

print("gamma_korr =", ugamma_1)
print("T_0_korr =", absolute_zero_1)
```

Die Ergebnisse sind in folgender Tabelle dargestellt:

| Größe | Messwert |
|---------|---------|
| $\gamma^{(0)}$  [$K^{-1}$] | $0.00367 \pm 0.00005$ |
| $\gamma^{(1)}$ [$K^{-1}$] | $0.00365 \pm 0.00005$ |
| $\vartheta_0^{(0)}$ [°C] | $-272.3 \pm 3.5$ |
| $\vartheta_0^{(1)}$ [°C] | $-274.0 \pm 4.0$ |
| $\vartheta_0^{\text{ref}}$ [°C] | $-273.15$ |

Der korrigierte Wert ist etwas näher am definierten Wert für den absoluten Nullpunkt. Für beide Messungen gilt, dass der Referenzwert innerhalb der Unsicherheiten ist. Dies zeigt, dass die oben angemerkten Effekte tatsächlich sehr klein sind und diese Methode zur groben Einschätzung des absoluten Nullpunktes geeignet ist.

Unsicherheiten wurden wiefolgt angenommen:

$\Delta h$ = 0.2 cm, $\Delta T$ = 0.1 $\vartheta$, $\Delta p_{norm}$ = 200 Pa

Die Unsicherheiten scheinen etwas zu konservativ gewählt zu sein, da die Unsicherheitsbereiche relativ groß sind im Gegensatz zu der Abweichung der Messwerte zu dem Referenzwert.

### Aufgabe 2.1: Messung nach der Methode von [Clément-Desormes](https://de.wikipedia.org/wiki/Experiment_von_Cl%C3%A9ment-Desormes) 

 * Bestimmen Sie $\kappa$ nach der **Methode von Clément-Desormes**.

**V E R S U C H S B E S C H R E I B U N G**

Zur Bestimmung des Adiabatenexponenten von Luft $\kappa$ wurde zuerst die Methode nach Clément-Desormes verwendet. Der Versuchsaufbau ist unten in einer Skizze gezeigt. Hierzu wurde in dem geschlossenen Kolben K zunächst mit einer Handpumpe ein geringer Überdruck erzeugt. Nach dem Abkühlen auf Raumtemperatur stellte sich der Zustand $(p_1,V_0,T_0)$ ein. Die zugehörige Höhendifferenz $\Delta h_1$ wurde am Flüssigkeitsmanometer abgelesen.

Anschließend wurde das Belüftungsventil für wenige Sekunden geöffnet und sofort wieder geschlossen. Es wurden insgesammt 3 Messreihen durchgeführt, bei denen das Belüftungsventil unterschiedlich lange geöffnet wurde (1s/3s/5s), um den Effekt von verschiedenen Offnungszeiten zu testen. Durch die schnelle Expansion des Gases erfolgte ein näherungsweise adiabatischer Druckausgleich mit der Umgebung, wodurch sich das Gas abkühlte. Nach dem Schließen des Ventils erwärmte sich das Gas bei konstantem Volumen wieder auf Raumtemperatur. Dabei entstand erneut ein kleiner Überdruck, der durch die Höhendifferenz $\Delta h_2$ bestimmt wurde.

Aus den gemessenen Höhendifferenzen lässt sich der Adiabatenexponent berechnen:

$$
\kappa = \frac{\Delta h_1}{\Delta h_1-\Delta h_2}.
$$

Um den statistischen Fehler so gering wie möglich zu halten, wurden mehrere Messwerte pro Belüftungszeit aufgenommen. Bei einer Belüftungszeit von 1s wurden 5, bei 3s wurden 10 und bei 5s wurden 5 Datenpunkte erhoben. Es wurde besonders darauf geachtet, dass möglichst ein Gleichgewichtszustand vor dem Ablesen des Manometers eingetroffen war.

---

```python
from os import getcwd
PWD=getcwd()

import matplotlib.pyplot as plt
import matplotlib.image as mpimg

img = plt.figure("demo", figsize=(10.0, 10.0))
img = mpimg.imread("Clement-Desormes.png".format(PWD=PWD))
imgplot = plt.imshow(img)
imgplot.axes.get_xaxis().set_visible(False)
imgplot.axes.get_yaxis().set_visible(False)
plt.show()
```

```python
import numpy as np
from uncertainties import ufloat
from uncertainties import unumpy as unp
# 3s
udelta_h1_3 = unp.uarray([110, 114, 210, 135, 105, 145, 125, 133, 191, 88], 2)#mm
udelta_h2_3 = unp.uarray([21, 30, 53, 33, 24, 34, 28, 30, 51, 21], 2)

# 1s
udelta_h1_1 = unp.uarray([131, 174, 166, 132, 104], 2)#mm
udelta_h2_1 = unp.uarray([77, 91, 93, 74, 50], 2)      

# 5s
udelta_h1_5 = unp.uarray([127, 160, 121, 104, 149], 2)#mm
udelta_h2_5 = unp.uarray([25, 32, 24, 17, 29], 2)


uarr_kappa_3 = udelta_h1_3/(udelta_h1_3 - udelta_h2_3)
uarr_kappa_1 = udelta_h1_1/(udelta_h1_1 - udelta_h2_1)
uarr_kappa_5 = udelta_h1_5/(udelta_h1_5 - udelta_h2_5)

#kappa_3, kappa_1, kappa_5 = np.mean(arr_kappa_3), np.mean(arr_kappa_1), np.mean(arr_kappa_5)
#kappa_3_std, kappa_1_std, kappa_5_std = np.std(arr_kappa_3), np.std(arr_kappa_1), np.std(arr_kappa_5)

weights_3 = 1 / unp.std_devs(uarr_kappa_3)**2
weights_1 = 1 / unp.std_devs(uarr_kappa_1)**2
weights_5 = 1 / unp.std_devs(uarr_kappa_5)**2

mean_3 = np.sum(weights_3 * unp.nominal_values(uarr_kappa_3)) / np.sum(weights_3)
mean_1 = np.sum(weights_1 * unp.nominal_values(uarr_kappa_1)) / np.sum(weights_1)
mean_5 = np.sum(weights_5 * unp.nominal_values(uarr_kappa_5)) / np.sum(weights_5)

print(f"mean (1s): {mean_1:.5f} +/- {np.sqrt(1/np.sum(weights_1)):.5f}")
print(f"mean (3s): {mean_3:.5f} +/- {np.sqrt(1/np.sum(weights_3)):.5f}")
print(f"mean (5s): {mean_5:.5f} +/- {np.sqrt(1/np.sum(weights_5)):.5f}")
```

Die Ergebnisse für die unterschiedlichen Belüftungszeiten werden in folgender Tabelle dargestellt:

| Belüftungszeit [s] | Messwert für $\kappa$ |
|---------|---------|
| 1 | $2.16\pm 0.03$ |
| 3 | $1.318 \pm 0.008$ |
| 5 | $1.23 \pm 0.01$ |

$\kappa$ wurde durch einen gewichteten Mittelwert bestimmt.

Wenn man die ermittelten Messwerte mit dem Referenzwert für den Adiabatenkoeffizient von Luft bei 20°C vergleicht (Wikipedia: $\approx$ 1.40), kann man erkennen, dass die Messungen mit einer Belüftungszeit von 3 und 5 Sekunden recht nah dran sind. Die Messung bei einer Belüftungszeit von 1s weicht stark von dem Referenzwert ab und wird daher nicht zur Bestimmung des Adiabatenkoeffizienten empfohlen.

Allgemein sind die Unsicherheiten sehr klein und damit ist der Referenzwert auch nicht innerhalb der Unsicherheiten. Das liegt zum einen daran, dass zusätzliche Effekte zur Verfälschung der Messwerte beitragen. Zum Beispiel könnte vor Ablesen des Drucks noch kein stationärer Zustand vorherrschen und damit würde sich die Druckanzeige eigentlich noch ändern. Dieser Effekt wird als Hauptursache für die starke Abweichbung eingeschätzt. Des Weiteren wurde die Unsicherheit für die Abschätzung der Höhen $\Delta h$ auf 2 mm abgeschätzt, was womöglich zu wenig konservativ war. Dadurch dass im Messaufbau ein Abstand zwischen Meterstab und Rohr ist, können Effekte wie Parallax die Längenabschätzung stärker verfälschen.

### Aufgabe 2.2: Messung nach der Methode von [Rüchardt](https://de.wikipedia.org/wiki/R%C3%BCchardt-Experiment)

 * Bestimmen Sie $\kappa$ nach der **Methode von Rüchardt**.

**Für diese Aufgabe stehen Ihnen zwei Messaufbauten zur Verfügung. Sie können zwischen einem Aufbau nach der originalen Methode nach Rüchard und einem Aufbau mit elektronischer Datenaufzeichnung und Kolbenprober wählen. Das Messprinzip ist für beide Aufgaben gleich. Nach der originalen Methode von Rüchard bestimmen Sie $\kappa$ nur für Luft. Mit dem Aufbau mit elektronischer Datenaufzeichnung und Kolbenprober bestimmen Sie $\kappa$ für Luft und für das Edelgas Argon (Ar).**

</div>

**V E R S U C H S B E S C H R E I B U N G**

Zur Bestimmung des Adiabatenexponenten $\kappa$ wurde der Aufbau mit elektronischer Datenaufnahme und Kolbenprober verwendet. Im Gegensatz zur ursprünglichen Methode nach Rüchardt kommen hierbei kein Glasrohr und keine frei bewegliche Kugel zum Einsatz. Stattdessen schwingt ein Kolben auf einem eingeschlossenen Gasvolumen, das bei der Bewegung des Kolbens abwechselnd adiabatisch komprimiert und expandiert wird.

Am Kolben befindet sich ein Magnet, dessen Bewegung durch eine um den Zylinder gewickelte Induktionsspule erfasst wird. Die induzierten Signale werden von einem Frequenzzähler registriert, sodass die Schwingungsperiode $T_{\Omega}$ elektronisch bestimmt werden kann. Durch Verschieben der Induktionsspule entlang des Zylinders wurde das eingeschlossene Gasvolumen $V$ variiert. Für sechs Volumina zwischen $30\,\mathrm{ml}$ und $80\,\mathrm{ml}$ wurden jeweils die zugehörigen Schwingungsperioden gemessen.

Die Messung wurde zunächst mit Luft und anschließend mit Argon durchgeführt. Für jedes eingestellte Volumen wurde die Schwingungsperiode aufgezeichnet. Unter der Annahme kleiner Auslenkungen des Kolbens ergibt sich für den Adiabatenexponenten

$$
\kappa = \left(\frac{2\pi}{T_{\Omega}}\right)^2 \frac{mV}{pA^2},
$$

wobei $m$ die Masse des Kolbens, $A$ die Querschnittsfläche des Zylinders, $V$ das eingeschlossene Gasvolumen und $p$ der Gasdruck sind. Aus den gemessenen Perioden wurde für Luft und Argon jeweils ein Wert für $\kappa$ bestimmt und anschließend mit den Literaturwerten verglichen.

---

```python
import matplotlib.pyplot as plt

um = ufloat(0.1098, 0.005)
up = ufloat(100000, 200)

ud = ufloat(0.0311, 0.001)
uA= np.pi * (ud/2)**2

# Luft
uV = unp.uarray([30, 40, 50, 60, 70, 80], 2)*0.001*0.001#m^3
uf_1 = unp.uarray([26.89, 21.76, 19.54, 17.98, 16.65, 15.57], 0.2)#Hz

# Argon
uf_2 = unp.uarray([25.97, 22.23, 20.11, 18.04, 16.67, 15.72], 0.2)#Hz

# berechnung k luft
uk_valuesLuft = []
for V_i, f_i in zip(uV, uf_1):
    uk_i = (2*np.pi * f_i)**2 * V_i* um / (up*uA**2)
    uk_valuesLuft.append(uk_i)

print("k Luft:" +str(uk_valuesLuft))

# plot 
plt.figure(figsize=(10, 6))
plt.errorbar(unp.nominal_values(uV)* 1000000, unp.nominal_values(uk_valuesLuft), yerr=unp.std_devs(uk_valuesLuft), fmt='o', capsize=5)
plt.xlabel("Messung mit Volumen (in ml)")
plt.ylabel("$\kappa$")
plt.title("Adiabatenkoeffizient $\kappa$ für Luft")
plt.show()
```

```python
# berechnung k argon
uk_values_argon = []
for V_i, f_i in zip(uV, uf_2):
    uk_i = (2*np.pi * f_i)**2 * V_i* um / (up*uA**2)
    uk_values_argon.append(uk_i)

print("k Argon:" +str(uk_values_argon))

# plot 
plt.figure(figsize=(10, 6))
plt.errorbar(unp.nominal_values(uV)* 1000000, unp.nominal_values(uk_values_argon), yerr=unp.std_devs(uk_values_argon), fmt='o', capsize=5)
plt.xlabel("Messung mit Volumen (in ml)")
plt.ylabel("$\kappa$")
plt.title("Adiabatenkoeffizient $\kappa$ für Argon")
plt.show()
```

In den Grafiken oben wurden die Messergebnisse dargestellt. Dabei wurden die gemessenen Werte für $\kappa$ jeweils gegen die zur Messung verwendeten Volumina geplottet. Bei der Messung mit Luft bei einem Volumen von 30ml kann man eine starke Abweichung vom Rest erkennen. Da deshalb die Wahrscheinlichkeit hoch ist, dass dieser Wert durch einen Messfehler hervorgerufen wurde, wird dieser Wert bei der ermittlung des Mittelwerts vernachlässigt.

Die Messreihe mit Argon scheint ohne Ausreißer zu sein.

```python
# Berchnung des gewichteten Mittelwerts
weights_Luft = 1 / unp.std_devs(uk_valuesLuft[1:])**2   # erster messwert raus
weights_Argon = 1 / unp.std_devs(uk_values_argon)**2

mean_Luft = np.sum(weights_Luft * unp.nominal_values(uk_valuesLuft[1:])) / np.sum(weights_Luft)
mean_Argon = np.sum(weights_Argon * unp.nominal_values(uk_values_argon)) / np.sum(weights_Argon)

print(f"mean (Luft): {mean_Luft:.2f} +/- {np.sqrt(1/np.sum(weights_Luft)):.2f}")
print(f"mean (Argon): {mean_Argon:.2f} +/- {np.sqrt(1/np.sum(weights_Argon)):.2f}")
```

Mithilfe des gewichteten Mittelwerts wurden die Adiabatenkoeffizienten bestimmt:

$$
\kappa_{\text{Luft}} = 1.45 \pm 0.09, \, \kappa_{\text{Argon}} = 1.49 \pm 0.09
$$

Die aus der kinetischen Gastheorie ermittelten Werte sind abhängig von der Anzahl der Freiheitsgrade des jeweiligen Gases (Ar: f = 3; $O^2$ und N: f = 5)

$$
\kappa = \frac{f + 2}{f}.
$$

Da Luft haupsächlich aus Stickstoff und Sauerstoff besteht, ergibt sich

$$
\kappa_{\text{Argon}}^{\text{theo}} \approx 1.66, \,  \kappa_{\text{Luft}}^{\text{theo}} \approx 1.40.
$$

Der Messwert für Luft stimmt innerhalb seiner Unsicherheiten mit dem theoretischen Wert überein, wohingegen der Messwert von Argon etwas stärker abweicht. Wahrscheinlich wurden die Unsicherheiten zu wenig konservativ abgeschätzt.

Wenn man den besten Messwert des Adiabatenkoeffizienten aus 2.1 ($\kappa_{\text{Luft}} = 1.318 \pm 0.008$) mit diesem Messwert vergleicht, kann man sagen, dass die Methode mit dem Kolbenprober leicht bessere Ergebnisse liefert, da der Erwartungswert näher am Referenzwert bzw. am theoretischen Wert liegt. Der große Vorteil der Methode ist, dass die Messwertaufnahme automatisiert ist und somit geringere statistische Unsicherheiten auftreten, die durch menschliche Einflüsse entstehen. Zum Beispiel war die Einschätzung der Belüftungszeit sehr subjektiv vom Testenden abhängig. Auch die Beurteilung wann sich ein Gleichgewichtszustand im Kolben eingestellt hat, kann von Messung zu Messung variieren und ist daher nicht optimal im Experiment.

## Aufgabe 3: Dampfdruckkurve von n-Hexan

 * Bestimmen Sie den Verlauf der **Dampfdruckkurve** von n-Hexan zwischen Zimmertemperatur und $0^{\circ}\mathrm{C}$. 
 * Schätzen Sie aus Ihrer Messung die **molare Verdampfungswärme $Q_{\mathrm{M}}$** ab. 

</div>

**V E R S U C H S B E S C H R E I B U N G**

Zur Untersuchung der Dampfdruckkurve von n-Hexan wurde ein mit flüssigem und gasförmigem n-Hexan gefüllter Glaskolben in ein Wasserbad eingebracht. Der Dampfdruck des n-Hexans wurde über ein direkt angeschlossenes Hg-Manometer bestimmt. Die Höhendifferenz der Quecksilbersäulen wurde mithilfe eines Kathetometers gemessen. Zu Beginn des Versuchs wurde bei Raumtemperatur die Referenzhöhe $h_0$ bestimmt, indem beide Quecksilberkuppen nacheinander angepeilt und ihre Höhendifferenz abgelesen wurden.

Anschließend wurde die Dampfdruckkurve zunächst bei sinkender Temperatur aufgenommen. Dazu wurden dem Wasserbad schrittweise Eis zugegeben. Nach jeder Temperaturänderung wurde ausreichend lange gewartet, bis sich zwischen flüssiger und gasförmiger Phase des n-Hexans ein thermodynamisches Gleichgewicht eingestellt hatte. Währenddessen wurde das Wasserbad automatisiert umgerührt, um Temperaturgradienten zu vermeiden. Für verschiedene Temperaturen wurden die Temperatur des Wärmebads sowie die zugehörige Änderung der Höhendifferenz $\Delta h$ gemessen.

Nach Erreichen der niedrigsten Temperatur wurde eine zweite Messreihe bei steigender Temperatur aufgenommen. Hierzu wurde dem Wasserbad portionsweise Wasser mit Raumtemperatur zugegeben. Erneut wurden Temperatur und Höhendifferenz bestimmt. Durch die Aufnahme einer Heiz- und einer Kühlkurve konnte überprüft werden, ob die Messpunkte bei gleicher Temperatur übereinstimmen oder ob Abweichungen durch Übersättigung des Dampfes, verzögerte Kondensation oder Siedeverzug auftreten.

Die Druckmessung beruht darauf, dass der Dampfdruck des n-Hexans mit sinkender Temperatur abnimmt. Dadurch verändert sich die Höhendifferenz der Quecksilbersäulen im Manometer. Aus der gemessenen Änderung $\Delta h$ lässt sich der Dampfdruck bei der jeweiligen Temperatur bestimmen. Für die Auswertung wurde die Clausius-Clapeyron-Gleichung verwendet, welche den Zusammenhang zwischen Dampfdruck und Temperatur beschreibt. Unter der Annahme, dass das Dampfvolumen gegenüber dem Flüssigkeitsvolumen dominiert und der Dampf als ideales Gas betrachtet werden kann, ergibt sich die Beziehung

$$
p(T)=p(T_0)\exp\left[-\frac{Q_{\mathrm M}}{R}\left(\frac{1}{T}-\frac{1}{T_0}\right)\right].
$$

Durch Logarithmieren erhält man ein lineares Modell

$$
\ln\left(\frac{p (T)}{p (T_0)}\right)=

-\frac{Q_{\mathrm M}}{R}
\left(
\frac{1}{T}-\frac{1}{T_0}
\right),
$$

mit dessen Hilfe die molare Verdampfungswärme $Q_{\mathrm M}$ aus den Messdaten bestimmt werden kann.


---

```python
T0 = ufloat(24.9, 0.2)
h_links = ufloat(44.35, 0.1)* 0.01#in m
h_rechts = ufloat(34.9, 0.1)*0.01#in m
rho_Hg = ufloat(13545.9, 0.0005) # in kg/m^3
p_Hg_0 = ufloat(0.163, 0.001) # in pascal

g = ufloat(9.809599, 0.000034)

p0 =  p_Hg_0 + rho_Hg * g * (h_links - h_rechts)

# nur messung von rechts da annahmen das Hg inkompressibel
T_sinkend = unp.uarray([23.2, 21.6, 19.5, 16.9, 15.2, 13.3, 10.3, 8.2, 6.3, 4.4, 0.9], 0.2)
h_sinkend_rechts = unp.uarray([35.2, 35.5, 35.9, 36.1, 36.7, 37.15, 37.65, 38.05, 38.25, 38.5, 38.85], 0.1)*0.01

T_steigend = unp.uarray([6.1, 8.0, 10.0, 11.0, 12.0, 13.0, 14.0, 15.1, 16.0, 17.3, 18.1, 19.1, 20.2], 0.2)
h_steigend_rechts = unp.uarray([38.2, 37.9, 37.6, 37.5, 37.3, 37.1, 36.95, 36.8, 36.6, 36.4, 36.25, 36.1, 35.9], 0.1)*0.01

p_sinkend = p0 - 2*rho_Hg * g * (h_sinkend_rechts - h_rechts)
p_steigend = p0 - 2*rho_Hg * g * (h_steigend_rechts - h_rechts)

plt.figure(figsize=(10, 6))
plt.plot(unp.nominal_values(T_sinkend), unp.nominal_values(p_sinkend) *0.01, 'o-', label='Messung bei sinkender Temperatur')
plt.plot(unp.nominal_values(T_steigend), unp.nominal_values(p_steigend) *0.01, 's-', label='Messung bei steigender Temperatur')
plt.xlabel('Temperatur [°C]')
plt.ylabel('Druck [hPa]')
plt.title('Dampfdruckkurve n-Hexan')       
plt.legend()
plt.grid(True)
```

Die Dampfdruckkurven stimmen größtenteils überein. Es gibt einen größeren Ausreißer bei T = 16.9 °C in der Messreihe bei sinkender Temperatur. Wahrscheinlich wurde bei der Entnahme des Messpunktes nicht lange genug auf das Eintreten des Gleichgewichts gewartet. 

Im Folgenden werden beide Messreihen auf das oben genannte Modell angepasst und $Q_M$ betimmt. Allerdings wird hierbei der stark abweichende Datenpunkt entfernt.

```python
from kafe2 import config, XYContainer, Fit, XYFit,  Plot

# entferne abweichung (4. messwert)
T_3sinkend = np.delete(T_sinkend, 3)
p_3sinkend = np.delete(p_sinkend, 3)

R = 8.31446261815324 # J/(mol*K)

# Modell
def ln_model(f, Q, f_0):
    return -Q/R *(f/1000 - f_0)

f_0 = 1/(T0 + 273.15)
f_sinkend= 1/(T_3sinkend + 273.15)
f_steigend= 1/(T_steigend + 273.15)

left_side_sinkend = unp.log(p_3sinkend/p0)
left_side_steigend = unp.log(p_steigend/p0)

xy_data_sinkend = XYContainer(x_data=unp.nominal_values(f_sinkend)*1000, 
                      y_data=unp.nominal_values(left_side_sinkend)) 

xy_data_sinkend.add_error(
    axis='y', 
    err_val=unp.std_devs(left_side_sinkend)
)  
xy_data_sinkend.add_error(
    axis='x', 
    err_val=unp.std_devs(f_sinkend)
) 

fit_sinkend = Fit(data=xy_data_sinkend, model_function=ln_model)

# constrain f_0
fit_sinkend.add_parameter_constraint(name='f_0',   value=unp.nominal_values(f_0),   uncertainty=unp.std_devs(f_0))

fit_sinkend.do_fit()

# plot the fit
plot = Plot(fit_objects=fit_sinkend)

plot.x_label = '1/T [1/mK]' 
plot.y_label = 'ln(p/p0)'

plot.plot()
plt.show()


# steigende Messreihe:

xy_data_steigend = XYContainer(x_data=unp.nominal_values(f_steigend)*1000, 
                      y_data=unp.nominal_values(left_side_steigend)) 

xy_data_steigend.add_error(
    axis='y', 
    err_val=unp.std_devs(left_side_steigend)
)  
xy_data_steigend.add_error(
    axis='x', 
    err_val=unp.std_devs(f_steigend)
) 

fit_steigend = Fit(data=xy_data_steigend, model_function=ln_model)

# constrain f_0
fit_steigend.add_parameter_constraint(name='f_0',   value=unp.nominal_values(f_0),   uncertainty=unp.std_devs(f_0))

fit_steigend.do_fit()

# plot the fit
plot = Plot(fit_objects=fit_steigend)

plot.x_label = '1/T [1/mK]' 
plot.y_label = 'ln(p/p0)'

plot.plot()
plt.show()
```

Trotz Entfernen des abweichenden Messwertes passt die Anpassung des Modells an die Messreihe bei sinkender Temperatur (obere Grafik) deutlich schlechter zu dem Modell als die Messreihe bei steigender Temperatur (untere Grafik). Das kann man unter Anderem am $\chi^2/\alpha$ -Wert von 2.229 (oben) und 0.5372 (unten) sehen. Letzerer Wert spricht sogar für zu konservativ eingeschätzte Unsicherheiten. Kleinere Unsicherheiten würden den $\chi^2/\alpha$ -Wert der oberen Messreihe noch schlechter werden lassen. Man kann erkennen, dass bei beiden Messreihen der lineare Trend bei geringen Temperaturen gebrochen wird. Eine mögliche Ursache hierfür ist die Übersättigung des n-Hexan-Dampfes beim Abkühlen. Erfolgt die Temperaturänderung schneller als die Einstellung des thermodynamischen Gleichgewichts, kann die Kondensation verzögert einsetzen. Der Dampfdruck liegt dann vorübergehend über dem Gleichgewichtswert, wodurch systematische Abweichungen der Messpunkte entstehen. Dieser Effekt tritt insbesondere bei niedrigen Temperaturen auf und stellt eine wesentliche Fehlerquelle der Messung dar. Wahrscheinlich würde man einen besseren Wert für $Q_M$ bekommen, wenn man diese Werte auch weglässt.

Mit den durchgeführten Anpassungen wurden folgende Werte für $Q_M$ berechnet:

$$
Q_M^1 = 43.1 \pm 1.4 \, \text{kJ/mol}, \, Q_M^2 = 39.1 \pm 1.2 \, \text{kJ/mol}, 
$$

wobei $Q_M^1$ der Wert aus der sinkenden Messreihe und $Q_M^2$ aus der steigenden Messreihe ist. Der gegebene Referenzwert für $Q_M$ beträgt 28.85 kJ/mol. Beide ermittelten Parameter sind zwar in der selben Größenordnung wie der Referenzwert, aber weichen stark ab. Der Grund dafür ist höchstwahrscheinlich der oben angesprochene Bruch des linearen Trends bei geringen Temperaturen.
