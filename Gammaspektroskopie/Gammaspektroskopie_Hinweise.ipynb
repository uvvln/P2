### Aufgabe 2.3: Energie-Kalibration des Detektors

**Hinweise zum Ablauf:**

 * Nehmen Sie eine **Kalibration von MCA-Kanalnummern (oder Histogramm-Bins) auf bekannte Photonenergien $E_{\gamma}$** vor.
 * Nutzen Sie hierzu **alle Photopeaks, die Ihnen durch die Präparate zur Verfügung stehen und den Annihilationspeak von $_{11}^{22}\mathrm{Na}$**.
 * Dabei handelt es sich um fünf Datenpunkte.

**Hinweise zur Auswertung:**

 * Beschreiben Sie Ihr Vorgehen.
 * **Passen Sie an die entsprechenden Strukturen geeignete Modelle an**, aus denen Sie für jeden Punkt $i$ die Energie $E_{\gamma}^{(i)}$ extrahieren können: 
   * Für den Single-Escape und die Photopeaks bietet sich eine [**Normalverteilung**](https://de.wikipedia.org/wiki/Normalverteilung) an.
   * Für die Compton-Kante bietet sich die [**Error function**](https://en.wikipedia.org/wiki/Error_function) an.
   * Beide sind aufgrund der statistischen Natur der zugrundeliegenden Prozesse sehr gut motiviert.
 * Für die Anpassungen sollten Sie die Spektren bereits **untergrundbereinigt** haben. 
 * Für die Anpassungen können Sie jede Struktur jeweils isoliert in den Bereichen $\Delta\mathrm{CH_{i}}$ betrachten, die Sie in **Aufgabe 2.2** bestimmt haben.  
 * Aus allen Modellen sollten Sie einen **Erwartungswert und eine Standardabweichung** (für die spätere Verwendung in **Aufgabe 2.4**) bestimmen können.
 * Entsprechende Beispiele für die Anpassung entsprechender Modelle finden Sie [hier](https://gitlab.kit.edu/kit/etp-lehre/p2-praktikum/students/-/tree/main/Gammaspektroskopie/tools).
 * Beachten Sie, dass den eigentlichen Strukturen noch zusätzlich Ausläufer der Strukturen überlagerter Prozessen unterliegen können, die Sie ggf. mit Hilfe eines geeigneten **zusätzlichen Polynom-Anteils** in Ihrem Modell berücksichtigen sollten.
 * Fügen Sie Ihrem Protokoll eine entsprechende **Kalibrationskurve $K(\mathrm{CH}_{i},E_{\gamma,i})$** zu und passen Sie ein geeignetes Modell an diese Kurve an. 
   * Sie können grundsätzlich von einem **linearen Verlauf der Kalibrationskurve** ausgehen.
   * Testen Sie auf das Vorhandensein von **Nicht-Linearitäten**, wie folgt:
     * Passen Sie ein geeignetes Polynom höherer Ordnung an die Datenpunkte an.
     * Überprüfen Sie, der $\chi^{2}$ normiert auf die Freiheitsgrade der Anpassung sich verbessert.
     * Überprüfen Sie, ob die entsprechenden Parameter im Rahmen ihrer aus der Anpassung bestimmten Unsicherheiten, mit Null verträglich sind.
 * Fügen Sie Ihrem Protokoll eine entsprechende Diskussion einschließlich aller ermittelten numerischen Werte und der $\chi2$-Werte normiert auf die Freiheitsgrade bei.
